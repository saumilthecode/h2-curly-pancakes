# Functional Abstraction

<!-- **Function** is a group of related statements that perform a specific task. -->

### Why use function?
* Functions help break program into smaller and modular chunks.
* It supports code reusability and reduces repeative code.
* It make large program more organized and manageable.


### General form of a function

```
def <name>(<formal parameters>):
	<body>
```

* Keyword `def` marks the start of a function.
* Every function has a name as uniquely identification.
* It may take in optional values, known as **parameters**, before running the code block.
* It may return value as a result using the `return` statement.
* All statements must be equally indented, which is usually 4 spaces. 

**Note:** Do not indent your program with mix of spaces and tabs.

Here is an example of a function:

````python
# def square(x): 
#     result = x * x 
#     return result

# print(square(2))
# print(square(2 + 5))
# print(square(square(3)))

def area_rect(x,y):         # Do not modify
    return x*y



def area_square(a):
    if a < 0:
        return 0
    else:
        return a*a

area_square(5)
````

It can be observed that the function `square(x)` would return the square of the parameter `x` inputted.
The same outputs can obtained using a different implementation, as shown below:

````python
def square(x): 
    result = x ** 2 
    return result

print(square(2))
print(square(2 + 5))
print(square(square(3)))
````

Hence, we can imagine a function like a black box which does something for us and we do not need to know how it does the job. This is what we know as **Functional Abstraction**.

In functional abstraction, details of the algorithms to accomplish the function are not visible to the user of the function. The user only needs to know the correct calling convention and to have trust in the accuracy of the functional results.

### Function with no input parameter nor return statement
* The following function simply prints "Hello World" when called.
* When there is no `return` statement used within a function, it is equivalent to `return None`.

````python
def hello_world_print():
    print("Hello World")

print('Running function..')    
hello_world_print() #this will print 'Hello World'
print()

a = hello_world_print()
print('Attempting to assign function call to variable a, before printing a')

print(a) #will this print 'Hello World'?
````

````python
def hello_world_return():
    return("Hello World")

a = hello_world_return()
print('Attempting to assign function call to variable a, before printing a')

print(a) #will this print 'Hello World'?
````

### Function with input parameter
* The following function takes an input value `who`, and prints "Hello {who}".
* Notice that since `print()` has been used within the function, there is no need to use it again in the test cases.

````python
def hello(who):
    print("Hello " + who)
    
hello('Adam')
hello('Eve')
hello('Kitty')
````

### Function with both input parameter and return statement
* The following function takes two input values and returns the greater value.

````python
def my_max(x, y):
    if x > y:
        return x
    else:
        return y
    
print(my_max(10,20))
print(my_max(-10,-20))
print(my_max(5,5))
````

# Variable Scope - Global vs Local

The scope of a variable determines the portion of the program where you can access a particular variable.
<br>
There are two basic variable scopes in Python:
* **Global variables:** variables defined **outside** a function body 
* **Local variables:** variables defined **inside** a function body

Global and Local variables are stored in different scopes:
<br>
1) Code in the global scope cannot use any local variables.
<br>
2) A local scope can access global variables.
<br>
3) Code in a function’s local scope cannot use variables in any other local scope.
<br>
4) You can use the same name for different variables if they are in different scopes.

You may want to run the following codes to understand the above points.

````python
# 1) Code in the global scope cannot use any local variables.
x = 4
def some_function(a,b):
  
    def square(x):
        result = x ** 2     # x is a local variable in the function square
        return result
  
    result = x + a + b      # x has not been defined in the global scope
    return result

some_function(3,4)
````

````python
# 2) A local scope can access global variables.

def another_function(a,b):
    k = 5                  # k is a global variable
 
    def square(x):
        result = x + k     # k can be used in the local scope of the function square
        return result
  
    result = square(a)
    return result

another_function(3,4)
````

````python
# 3) Code in a function’s local scope cannot use variables in any other local scope.
y = 6
def my_function(a,b):
    def square(x):
        result = x + y        # y has not been defined in the global scope nor the local scope of function square
        return result
    def sqrt(y):
        result = y ** 0.5     # y is a local variable in the function sqrt
        return result
     
    result = square(a)
    return result

my_function(3,4)
````

````python
# 4) You can use the same name for different variables if they are in different scopes.

def hypothenuse(a,b):
    def square(x):
        result = x ** 2       # x is a local variable in the function square
        return result
    def sqrt(x):
        result = x ** 0.5     # x is also a local variable in the function sqrt
        return result
     
    result = sqrt (square(a) + square(b))
    return result

hypothenuse(3,4)
````
