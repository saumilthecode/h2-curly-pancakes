## Question 1 : Mathematic Computation

```python
def power(base, exp):
    if exp == 0:
        return 1
    return base * power(base,exp-1)

#Do not modify / remove the following lines
print(power(2, 0))
print(power(-3, 9))
```

## Question 2 : String Manipulation


```python
def reverse_string(s):
    if len(s) == 0:
        return ''
    return s[-1]+ reverse_string(s[:len(s)-1]) 




reverse_string('abcde')
```

## Question 3 : Sum of Elements in a List

```python
def list_sum(arr):
    if len(arr) == 0:
        return 0
    return arr[0] + list_sum(arr[1:])


list_sum([1,2,3])
```

## Question 4 : Maximum Element in a List

```python
def find_max(arr):
    if len(arr) == 1:
        return arr[0]
    
    maxx = find_max(arr[1:])
    if arr[0] > maxx:
        return arr[0]
    else:
        return maxx
     


# Do not modify the following:
find_max([2,5,3,9,1,0])
```

## Question 5 : Counting the Number of Students Passing a Test

```python
def count_passes(students, index):
    if len(students) == index:
        return 0
    if students[index][1] >= 50:
        return 1 + count_passes(students,index+1)
    else:
        return count_passes(students,index+1)



# Do not modify the following:
class1 = [('Adam', 30), ('Bob', 68), ('Chew Fun', 79), ('Daniel', 45), ('Edward', 50)]


count_passes(class1, 0)
```
