## Question 1

```python
# Write the code for the function validate here:
def isValid(NRIC):
    if len(NRIC) != 9:
        return False
    if NRIC[0] not in ('S', 'T'):
        return False
    if not NRIC[1:8].isdigit():
        return False
    if not NRIC[8].isalpha():
        return False

    return True


print(isValid('S1234567T') == True) 
print(isValid('RST12345T') == False)
```

## Question 2

Write program code for a number guessing game for numbers between 1 to 100.

```python
print("Number Guessing Game!")
print("You have 6 tries.")
ans = 37

for i in range(6):
    guess = input("Please enter a number (1-100):")
    if not guess.isdigit():
        print("Please enter only numbers, not any other character.")
        continue
    guess = int(guess)
    if guess < 1 or guess > 100:
        print("Please enter a number within 1 to 100.")
        continue
    if guess < ans:
        print("Your guess was too low!")
    elif guess > ans:
        print("Your guess was too high!")
    else:
        print("You are correct!")
        break
        
# Do not modify the last line of the code:

print("End of game")
```

## Question 3

Modify your Number Guessing Game code in the previous question to report a tuple containing all valid guesses at the end of the game.

```python
# Copy and paste your code from the previous 
# question in the space below:
print("Number Guessing Game!")
print("You have 6 tries.")
ans = 37

guesses = ()
for i in range(6):
    guess = input("Please enter a number (1-100):")
    if not guess.isdigit():
        print("Please enter only numbers, not any other character.")
        continue
    guess = int(guess)
    if guess < 1 or guess > 100:
        print("Please enter a number within 1 to 100.")
        continue
    guesses = guesses + (guess,)
    if guess < ans:
        print("Your guess was too low!")
    elif guess > ans:
        print("Your guess was too high!")
    else:
        print("You are correct!")
        break

print("Your guesses were:", guesses)
        
# Do not modify the last line of the code:

print("End of game")
```

```python
food_prices = (
    ("Chicken Rice", 4.50),
    ("Nasi Lemak", 3.80),
    ("Laksa", 5.00),
    ("Char Kway Teow", 4.20),
    ("Roti Prata (2 pcs)", 2.00),
    ("Mee Rebus", 3.50),
    ("Wanton Mee", 4.00),
    ("Bubble Tea (M)", 3.90),
    ("Kaya Toast Set", 3.80),
    ("Western Breakfast Set", 6.50))


def find_price_of_food(foodstr):
    for itemset in food_prices:
        if foodstr.lower() in itemset[0].lower():
            return itemset
    return None


print(find_price_of_food("nAsi") == ('Nasi Lemak', 3.80))
print(find_price_of_food("prata") == ('Roti Prata (2 pcs)', 2.00))
```
