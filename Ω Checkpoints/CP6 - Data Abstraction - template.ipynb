## Question 2: Read data from Text file into Tuples  [5 marks]

```python
def read_data(filename):
    StudentRecord = []
    with open(filename) as f:
        lines = f.readlines()
        f.close()
    for line in lines[1:]:
        student_id, name, subject, score = line.strip().split(",")
        student = make_student(student_id, name, subject, score)
        StudentRecord.append(student)
    return tuple(StudentRecord)
        

results = read_data('students.txt')
print(results)
```

## Question 3: Compute the Average Score  [5 marks]

```python
def make_student(student_id, name, subject, score):
    return  (student_id, name, subject, score)
def get_id(student):
    return student[0]
def get_name(student):
    return student[1]
def get_subject(student):
    return student[2]
def get_score(student):
    return int(student[3])
def is_pass(student):
    if get_score(student)>= 50:
        return True
    else:
        return False
####
def average(data, subject):
    sum = 0
    count = 0
    for item in data:
        if get_subject(item) == subject:
            sum += get_score(item)
            count +=1
        else:
            pass
    if count == 0:
        return None
    else:
        aver = sum / count
        return float(round(aver,2))
    
            

results = (('S101', 'Alice', 'Computing', 85), ('S102', 'Bob', 'Mathematics', 62), ('S103', 'Charlie', 'Computing', 45), ('S104', 'Diana', 'Physics', 91), ('S105', 'Evan', 'Computing', 78))
average(results, 'Computing')
```

## (Extension) Question 4: Display the Average Scores for each Subject  [5 marks]

```python
def display(data):
    subjects = []

    for student in data:
        subject = get_subject(student)
        if subject not in subjects:
            subjects.append(subject)

    print(f"{'Subject':<15}{'Average Score':<15}")

    for subject in subjects:
        print(f"{subject:<15}{average(data, subject):<15}")

display(results)
    
```
