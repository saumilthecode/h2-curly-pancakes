```python
def f(x):
    if x <2:
        return x
    else:
        if x %2 == 0:
            return f(x/2)
        else:
            return 1+f(x-1)
```

```python
def power_10(k):
    if k <10:
        return 0
    else:
        return 1 + power_10(k*0.1)
   

def significand(k):
    if k <10:
        return round(k,2)
    else:
        return significand(k*0.1)


power_10 (123000)
significand(123)
```

```python
def sum_num(n):
    if n >0:
        if n %2 == 0 or n %3 == 0:
            return n + sum_num(n-1)
        else:
            return sum_num(n-1) 
    else:
        return 0
        
sum_num(7)
```

```python
def inception(n):
    if n == 0:
        return ""
    elif n ==1:
        return 'Dream'
    else:
        return 'Dream within a '+ inception(n-1)
        
inception(2)
```

```python
def g(x):
    if x < 5:
        return x
    else:
        return  (g(x-1) + g(x-2)) -2
g(5)
```

```python
def swap_char(string):
    if len(string) == 1:
        return string
    else:
        if ord(string[0]) < ord(string[1]):
            return string[0] + swap_char(string[1:])
        else:
            return string[1] + swap_char(string[0] + string[2:])
swap_char("xy")
```
