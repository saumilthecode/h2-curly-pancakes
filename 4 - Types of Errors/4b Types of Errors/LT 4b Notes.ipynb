# Types of Error

In order for programs to run correctly or as intended, careful planning and testing of a program is essential. When developing programs, it is quite common to encounter error messages.

The following examples illustrate some of these errors. Try running them to identify the different types of errors.

```python
# Omitting return statement
def square(x):
    x * x

square(2)
```

```python
# Incompatible types
def square(x):
    return x * x

print('Answer is : ' + square(5))
```

```python
# Incorrect no. of  arguments
def square(x):
    return x * x
    
square(3,5)
```

```python
# Syntax error
def double(x):
    return 2x 

print(double(3))
```

```python
# Arithmetic error
x = 3
y = 0

x/y
```

```python
# Undeclared variables
x = 2
x + k
```

```python
# Infinite loop (to be covered in later LTs)
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)

factorial(-1)
```

```python
# Numerical imprecision (floating point error)
((10)**(1/2))**2
```

```python
# Logic error

def check(x):
    if x > 100:
        return 'Big'
    elif x > 2000:
        return 'Very Big!'
    
print(check(101))
print(check(2001))
```

# Debugging

Debugging is the process of detecting and removing of existing and potential errors (also called as ‘bugs’) in a program code that can cause it to behave unexpectedly or crash. To prevent incorrect operation or system to run, debugging is used to find and resolve bugs or defects. Sometimes it takes more time to debug a program than to code it.

For example, try to understand the following code below and predict the output for `f(6)` before running it. Does your prediction match the output? 

```python
x = 5
y = 10
z = 15
def f(x):
    def g(y):
        def h(z):
            return x + y + z
        return h(y)
    return g(x)

f(6)
```

To understand what was happening, sometimes adding **print statements** at appropriate lines would be helpful. In fact, print statements can often be used in debugging our program codes.

Try running the following codes with print statements, do they help you to understand why the code produced such an output?

```python
x = 5
y = 10
z = 15
def f(x):
    print('1. x =', x, ', y =', y, ', z =', z)
    def g(y):
        print('2. x =', x, ', y =', y, ', z =', z)
        def h(z):
            print('3. x =', x, ', y =', y, ', z =', z)
            return x + y + z
        return h(y)
    return g(x)

f(6)
```

# Test Cases

A test case is a test scenario measuring functionality across a set of program codes or conditions to verify the expected result. It usually consists of an input to be entered and an expected output to verify a program's actual output against its expected output. If the program's output matches the expected output, we can say that the program passes that particular test case.

Test cases are supposed to be chosen meaningfully. Some appropriate test cases to be considered are:
- Normal Data Values
- Extreme / Boundary Data Values
- Abnormal Data Values
- Volume Data Values

For example, the following code is being used to compute the percentage based on the given `score` and `total`.

```python
def percentage(score,total):
    return (score/total)*100

# Normal Data Values
print(percentage(20,80) == 25.0)
print(percentage(30,60) == 50.0)
```

```python
# Extreme / Boundary Data Values
print(percentage(0,80) == 0.0)
print(percentage(60,60) == 100.0)
```

```python
# Abnormal Data Values
print(percentage(-10,80) == 'Error')
print(percentage(120,60) == 'Error')
```

Notice that the program fails to work as expected in the last two test cases. This is because we actually do not expect the `score` to be negative or to be more than the `total`. In order to handle these invalid values, we can include some conditional checks as shown below.

```python
def percentage(score,total):
    if score < 0 or score > total:
        return 'Error'
    else:
        return (score / total) * 100

# Abnormal Data Values
print(percentage(-10,80) == 'Error')
print(percentage(120,60) == 'Error')
```

For programs which reads large amount of data, we may need to input multiple or large amount of data. A testing done on such high volumes of data is called the **volume testing**. It is usually done to test if the program is efficient (has reasonable response time) with increasing volume of data being provided.

```python
# Modify the code below and make it work:
# divisible by 400 or a year that is not divisible by 100 if it's divisible by 4.

def test_leap_year(year):
    if year % 400 == 0 or not (year % 4 == 0 and year % 100 == 0):
        return 'Leap year!'
    else:
        return 'Not a leap year!'

print(test_leap_year(2004)) #is a leap
print(test_leap_year(1904)) #is a leap
print(test_leap_year(2014)) # not a leap
```
