# Debugging Exercise 2
## Inserting print statements

Let us examine the following code for finding the solutions of a quadratic equation. The values of the coefficients *a*, *b*, and *c* are given at the beginning of the program.

```python
from math import *

def quadratic (a, b, c):

    delta = b * b - 4 * a * c

    s1 = (-b - sqrt(delta)) / (2 * a)
    s2 = (-b + sqrt(delta)) / (2 * a)

    return s1, s2
```

Try running the following code to find the solutions to the quadratic equation $x^2+2x-3=0$. You should be able to obtain two solutions.

```python
quadratic(1,2,-3)
```

Try running another test case :

```python
quadratic(548,753,162784)
```

## Tracing the call stack

We see that the error occurs at line 7 and the message is "**ValueError:** math domain error". This message may seem confusing. If you are brave enough, take out a piece of paper or use a calculator to check the solution.

Let us read our code again. In line 5, we compute `delta`. We know that if this value is negative, the squareroot is undefined. Is the value of `delta` negative in our code? Let's check by printing the value of delta just after line 5.

Note that the Python interpreter executes your program line by line, and stop immediately when an error is encountered. Therefore even if you add the print at the end of the program, you won't be able to see the value of `delta` because when the interpreter encounters the error, it has not reached the print statement yet.

Now examine the value of `delta`. Is it negative?

```python
from math import *

def quadratic (a, b, c):

    delta = b * b - 4 * a * c
    print(delta)
    s1 = (-b - sqrt(delta)) / (2 * a)
    s2 = (-b + sqrt(delta)) / (2 * a)

    return s1, s2

quadratic(548,753,162784)
```

This clarifies the error message which you have gotten. The `math.sqrt()` function does not know how to handle a negative value, hence the math domain error message. At this point, we are quite sure that the error is caused by the negative value of `delta`. (Remember when you input $a=1, b=2, c=-3$, the program works because the `delta` value was positive.)

## Fixing the bug

We can easily fix this program with an `if` statement. For example, if `delta` is negative, print a line which says that there is no solution. If it is positive (or zero), return the two (or one) solutions of the quadratic equation.
