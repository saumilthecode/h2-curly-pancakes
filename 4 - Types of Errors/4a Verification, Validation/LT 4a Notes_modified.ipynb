# Data Verification

**Data verification** is a way of preventing errors when data is copied from one medium to another.

Verification does not check if data makes sense or is within acceptable boundaries, it only checks that the data entered is identical to the original source.

Two most commonly used verification methods are:
- Double entry - Data is entered twice and the computer checks that they match up
- Proof reading (visual check) - The user manually reads and compares the newly inputted data against the original source to ensure they match

See https://en.wikipedia.org/wiki/Data_verification

# Data Validation

**Data validation** is the process of ensuring data have undergone data cleansing to ensure they have data quality, that is, that they are both *correct* and *useful*.

It uses routines, often called "validation rules", "validation constraints" or "check routines", that check for correctness, meaningfulness, and security of data that are input to the system.

See https://en.wikipedia.org/wiki/Data_validation

## Different Kinds of Data Validation

### 1.  Presence Check

This validation checks that required data has been input by the user and the system will reject groups of data where required fields have been left blank.

This is often used with data collection forms, as shown in the following example:

```python
item = input('Insert text value: ')

while item == '':
    item = input('Please input text value: ')
```

In the above example, a `WHILE` loop is used to check for the presence of the input. If the data is missing, then the user will be prompted again to enter the required data.

### 2. Range Check

In a range validation, user's input is checked for consistency with a minimum/maximum range. Data entered that falls outside the expected range will be rejected.

For example, a Singapore mobile number must have 8 digits and the first digits must be either 8 or 9.

```python
n = int(input('Enter a Singapore mobile number: '))

if  79999999 < n < 100000000:
    print('The mobile number is', n)
else:
    print('Invalid mobile number.')
```

### 3. Length Check

A length validation would check that the data entered is of a reasonable length. Data items that have a length outside the expected values are rejected.

For example, a password entered must have more than six characters.

```python
password = input('Enter your password: ')

while len(password) < 6:
    password = input('Password entered must have six or more characters. Re-enter your password: ')
```

### 4.  Type Check

The data type validation checks that the data provided through user input are consistent with the expected data type. It will reject any input that is of a different type.

Since Python's `input()` function will always return a String data type, we need to use another function called `eval()` in order to return the appropriate type. Try to run the following code with inputs from different data types:

```python
input_type = input('Input data of a particular data type: ')
print(type(input_type))

input_type = eval(input_type)
print(type(input_type))
```

We can apply the above information to require a user to input an integer.

We will use some_string.isnumeric() to check if the string contains a number representation.

```python
number = input('Please input an integer: ')

while number.isnumeric() is False:
    number = input('You have not entered an integer, please try again: ')
```

### 5. Format Check

A format validation checks that a data item matches a predetermined pattern and that the individual elements of the data item have particular values or data types.

The following example illustrates the validation of an email address:

```python
email = input('Enter a valid ICON email address: ')
index = email.find('@')

if email[index:] == '@students.edu.sg':
    print('The email address is:', email)
else:
    print('Invalid MOE email address')
```
