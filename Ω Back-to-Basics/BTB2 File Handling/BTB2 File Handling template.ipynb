### Q1: Reading from a Text file
Write the program code to read the students' data from the given text file into a tuple of tuples. Each tuple should contain the student's name, gender, height and weight.

```python
def openfile(filename):
    f = open(filename)
    rettup = ()
    for line in f:
        data = line.strip().split('\t')
        name = data[0]
        gender = data[1]
        height = float(data[2])
        weight = float(data[3])
        rettup += ((name, gender, height, weight),)
    f.close()
    return rettup

data_tuple = openfile('data.txt')
print(data_tuple)
```

### Q2: Compute average height and weight

```python
def average(tup):
    sumheight = 0
    sumweight = 0
    for tupl in tup:
        sumheight += tupl[2]
        sumweight += tupl[3]
    return (round(sumheight/len(tup),2),round(sumweight/len(tup),2))
        




data_tuple = openfile('data.txt')
mean_values = average(data_tuple)

print(mean_values)
```
