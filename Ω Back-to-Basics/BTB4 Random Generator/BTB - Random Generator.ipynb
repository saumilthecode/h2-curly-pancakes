### 1. Generate Random Number - with and without using `seed()`.
- Run the following code twice without using `seed(30)` and observe the two lists of random numbers generated. 
- Run the following code twice with `seed(30)` and observe the two lists of random numbers generated. 

```python
from random import *

#seed(30)       # optional, can use other value.

lst = []

for i in range(10): 
    x = round(random(),2)
    lst.append(x) 
    
print(lst)
```

### How to generate floating numbers between 0 to 5 and in 2 decimal places? 

```python
from random import *

#seed(30)       # optional, can use other value.

lst = []

for i in range(10): 
    x = round(random() * 5,2)
    lst.append(x) 
    
print(lst)


```

### 2. Generate Random Integers
- Run the following code twice without using `seed(30)` and observe the two lists of random numbers generated. 
- Run the following code twice with `seed(30)` and observe the two lists of random numbers generated. 

```python
from random import *

#seed(30)       # optional, can use other value.

lst = []

for i in range(10): 
    x = randint(0, 9)
    lst.append(x) 
    
print(lst)
```

### 3a. Random selection from a list
- Select one random item from the list.
- Run the code multiple times to get different random outputs.

```python
from random import choice

lst = ['Adam', 'Bob', 'Charles', 'Daniel']

print(choice(lst))
```

### 3b. Random selection(s) from a list
- Select two random items from the list.
- Run the code multiple times to get different random outputs.

```python
from random import choice, choices, sample

lst = ['Adam', 'Bob', 'Charles', 'Daniel']

method1 = choices(lst, k=2)

method2 = sample(lst, 2)

print(method1)
print(method2)
```

### 4. Generate Normally Distributed Numbers (with mean and SD)
- The following code will generate 1000 normally distributed random numbers: 

```python
from random import *

lst = []
for i in range(1000):
    x = normalvariate(50,7)
    lst.append(x)

print(lst)    
```

### How to find the `mean` and `standard deviation` of the data in a list? 

```python
from statistics import *

print(mean(lst))
print(stdev(lst))
```

```python
from random import choice

def main():
    f = open("Namelist.TXT", "r")
    students = []
    for line in f:
        name = line.strip()
        if name != "":
            students.append(name)
    f.close()
    while True:
        if len(students) == 0:
            print("There are no more students left to pick.")
            return

        ans = input("Do you want to pick a random student's name? (yes/no): ")
        if ans.lower() == "yes":
            picked = choice(students)
            print(picked)
            students.remove(picked)

        elif ans.lower() == "no":
            return
        else:
            print("Please enter yes or no.")
```

```python
from random import randint

def generate(n):
    nums = []
    for i in range(n):
        nums.append(randint(1, 20))
    return nums
```

```python
def count(seq):
    dic = {}
    for i in seq:
        if i in dic:
            dic[i] += 1
        else:
            dic[i] = 1
    return dic
```

```python
from random import randint

freq = {}
for i in range(1000):
    num = randint(1, 20)

    if num in freq:
        freq[num] = freq[num] + 1
    else:
        freq[num] = 1


# Task 1.1
print("Integer, Frequency")

for num in range(1, 21):
    print(num, ":", freq[num])


# Task 1.2
print("Integer, Actual Freq, Difference")

expected = 50

for num in range(1, 21):
    actual = freq[num]
    difference = actual - expected
    print(num, ":", actual, difference)
```
