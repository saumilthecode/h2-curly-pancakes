## Coursemology Question 4: Position ADT

The constructor method to create a Position ADT uses a tuple to store the x- and y-coordinates of the position of a point.


```python
def make_point(xcoord, ycoord):
    return (xcoord, ycoord)


Point1 = make_point(0, 0)
print(Point1)
```

As a programmer, write the program codes for the following methods:
- `get_x(point)` to return the x-coordinate value
- `get_y(point)` to return the y-coordinate value
- `midpoint(point1, point2)` to return a Position ADT
- `gradient(point1, point2)` to return a float
- `distance(point1, point2)` to return a float
- `equation(point1, point2)` to return a string "y = mx + c"
- `display(point)` to return a string (x, y)


```python
def get_x(point):
    return point[0]

def get_y(point):
    return point[1]

def midpoint(point1, point2):
    totalx = get_x(point1)+get_x(point2) 
    totaly = get_y(point1)+get_y(point2) 

    return (totalx/2,totaly/2)
    
def gradient(point1, point2):
    run = get_x(point1)-get_x(point2) 
    rise = get_y(point1)-get_y(point2)
    return rise/run
    
def distance(point1, point2):
    run = get_x(point1)-get_x(point2) 
    rise = get_y(point1)-get_y(point2)
    return (run**2 + rise**2) **0.5

def equation(point1, point2):
    m = gradient(point1,point2)
    c = get_y(point1)-m*get_x(point1)
    return ('y = '+str(m)+'x'+ ' + ' + str(c))

def display(point):
    return (string(point))
```

```python
Point1 = make_point(0,0)
Point2 = make_point(2,2)

print(get_x(Point1))
print(get_y(Point1))
print(midpoint(Point1, Point2))
print(equation(Point1, Point2))
```

## Coursemology Question 5: Rational Number ADT

The constructor method to create a Rational Number ADT uses a tuple to store the numerator and denominator of the rational number or fraction.


```python
def make_rat(numer, denom):
    return (numer, denom)


# Do not modify the following:
rat1 = make_rat(2,3)
rat2 = make_rat(5,6)

print(rat1)
print(rat2)
```

As a programmer, write the program codes for the following methods:
- `get_numer(rat)` to return the numerator of the Rational Number
- `get_denom(rat)` to return the denominator of the Rational Number
- `add(rat1, rat2)` to return a Rational Number ADT
- `sub(rat1, rat2)` to return a Rational Number ADT
- `mul(rat1, rat2)` to return a Rational Number ADT
- `div(rat1, rat2)` to return a Rational Number ADT
- `print_rat(rat)` to return a String "n/d"
- `is_equal(rat1, rat2)` to return a Boolean
- `is_whole(rat)` to return a Boolean

```python
def get_numer(rat):
    return rat[0]

def get_denom(rat):
    return rat[1]

def add(rat1, rat2):
    numer = get_numer(rat1) * get_denom(rat2) + get_numer(rat2) * get_denom(rat1)
    denom = get_denom(rat1) * get_denom(rat2)
    return (numer, denom)

def sub(rat1, rat2):
    numer = get_numer(rat1) * get_denom(rat2) - get_numer(rat2) * get_denom(rat1)
    denom = get_denom(rat1) * get_denom(rat2)
    return (numer, denom)

def mul(rat1, rat2):
    numer = get_numer(rat1) * get_numer(rat2)
    denom = get_denom(rat1) * get_denom(rat2)
    return (numer, denom)

def div(rat1, rat2):
    numer = get_numer(rat1) * get_denom(rat2)
    denom = get_denom(rat1) * get_numer(rat2)
    return (numer, denom)

def print_rat(rat):
    return str(get_numer(rat)) + "/" + str(get_denom(rat))

def is_equal(rat1, rat2):
    return get_numer(rat1) * get_denom(rat2) == get_numer(rat2) * get_denom(rat1)

def is_whole(rat):
    return get_numer(rat) % get_denom(rat) == 0

```

```python
# Do not modify the following:
x = make_rat(2,3)
y = make_rat(5,6)

b = make_rat(4,2)
c = make_rat(6,3)

print(is_whole(x))
print(is_whole(b))
print(is_equal(x, b))
print(is_equal(b, c))
```

## Coursemology Question 6: Reduce Rational Number to Lowest Term


```python
from math import gcd

def make_rat(n, d):
    divisor = gcd(n, d)
    return (n // divisor, d // divisor)

def get_numer(rat):
    return rat[0]

def get_denom(rat):
    return rat[1]

def add(rat1, rat2):
    numer = get_numer(rat1) * get_denom(rat2) + get_numer(rat2) * get_denom(rat1)
    denom = get_denom(rat1) * get_denom(rat2)
    return make_rat(numer, denom)

x = make_rat(2, 3)
y = make_rat(5, 6)

print(add(x, y))
```

## Coursemology Question 7: Sum all the Fractions


```python
rat1 = make_rat(1,2)
rat2 = make_rat(1,4)
rat3 = make_rat(2,7)
rat4 = make_rat(8,9)


tuplefrac = (rat1, rat2, rat3, rat4)  #tuplefrac contains 4 fractions



def sumfrac(fractions):
    totalrat = make_rat(0,1)
    for frac in fractions:
        totalrat = add(totalrat,frac)
    return totalrat
sumfrac(tuplefrac)
```
