# Coursemology Question 1: Read Data from a Text File

Write program code for the function `read_csv(filename)` to return a tuple of tuples containing the data from the text file.

Note: The header in the text file should be excluded.


```python
def read_csv(filename):
    file = open(filename)
    lines = file.readlines()
    data = ()
    for line in lines[1:]:
        line = line.strip()
        row = line.split(',')
        data = data + (tuple(row),)
    file.close()
    return data


computing_class = read_csv('data.csv')
print(computing_class)
```

# Coursemology Question 3: 

The tuple `computing_class` contains all the Student ADTs.

Each student ADT is a tuple containing info like the CG class, name, gender, XP's level and XP's score of the student. 

Write program code for the function `average_score(group)` and `average_level(group)` using following methods to compute the average score and level for the Computing class:

- `get_class(student)` -> string
- `get_name(student)`  -> string
- `get_gender(student)` -> string
- `get_level(student)` -> integer
- `get_score(student)` -> integer
- `size(group)` -> integer


```python
def average_score(group):
    totascore = 0
    for student in group:
        totascore += get_score(student)
    return round(totascore/size(group),2)
        

def average_level(group):
    totascore = 0
    for student in group:
        totascore += get_level(student)
    return round(totascore/size(group),2)



# Do not modify the following:
computing_class = read_csv("result.csv")

print(average_score(computing_class))
print(average_level(computing_class))
```

# Coursemology Question 4: 

The tuple `computing_class` contains all the Student ADTs.

Each student ADT is a tuple containing info like the CG class, name, gender, XP's level and XP's score of the student. 

Write program code for the function `class_list(group, CG)` using the following methods to return a tuple containing all the still from the same CG class in the `computing_class`.

- `get_class(student)` -> string
- `get_name(student)`  -> string
- `get_gender(student)` -> string
- `get_level(student)` -> integer
- `get_score(student)` -> integer
- `size(group)` -> integer


```python
def class_list(group, CG):
    stud_lst = ()
    for student in group:
        if get_class(student) == CG:
            stud_lst += (get_name(student),)
    return stud_lst

# Do not modify the following:
computing_class = read_csv("result.csv")

class_list(computing_class, '05S07')
```
