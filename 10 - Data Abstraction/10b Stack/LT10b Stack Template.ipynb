# LT10b Stack ADT

A Stack is a fundamental Abstract Data Type (ADT) that operates on a Last-In, First-Out (LIFO) principle. Imagine a stack of plates: you can only add a new plate to the top, and you can only remove the topmost plate. The last plate you put on is the first one you take off.


- Abstract Data Type (ADT): A stack is defined by its behaviour (what it does) rather than its implementation (how it does it). It specifies a set of operations without detailing the underlying data structures.

- LIFO (Last-In, First-Out): This is the defining characteristic. The last element added to the stack is the first element to be removed.

```python
# Q5
def peek(stack):
    if stack == None or len(stack) == 0:
        return None
    else:
        return stack[len(stack)-1]
    
def clear(stack):
    if stack == None:
        return 
    else:
        for i in range(len(stack)):
            stack.pop()
        return stack


##########################################
#       Do not modify test code          #              
##########################################
s1 = make_empty_stack()                  #
s2 = make_stack((2, 4, 5))               #
                                         #
peek1 = peek(s1) #None                   #
peek2 = peek(s2) #5                      #
clear(s2)                                #
peek3 = peek(s2) #None                                                            #
##########################################

print(s1)
print(s2)

print(peek1)
print(peek2)

clear(s2)
print(peek3)
```

```python
# Q6

def reverse(string):
    stack = make_empty_stack()
    for i in range(len(string)):
        push(stack, string[i])
    result = ""
    while is_empty_stack(stack) == False:
        result += pop(stack)
    return result
```

```python
# Q7
def denary_to_binary(n):
    stack = make_empty_stack()
    if n == 0:
        return "0"
    while n > 0:
        push(stack, n % 2)
        n //= 2
    binary = ""
    while not is_empty_stack(stack):
        binary += str(pop(stack))
    return binary
```

```python
# Q11
def calculate(inputs):
    stack = make_empty_stack()
    if len(inputs) == 1:
        return inputs[0]
    for char in inputs:
        if char=='+'or char == '-'or char == '*' or char== '/':
            rhs = pop(stack)
            lhs = pop(stack)
            if char == '+':
                push(stack, lhs + rhs)
            elif char == '-':
                push(stack, lhs - rhs)
            elif char == '*':
                push(stack, lhs * rhs)
            else:
                push(stack,lhs / rhs)
        else:
            push(stack, char)
    return pop(stack)
calculate((1, 2, '*', 3, '-', 2, '*', 5, '+'))
```
