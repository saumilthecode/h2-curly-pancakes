# Question 1

```python
def hash(string):
    total = 0
    for i in range(len(string)):
        total += ord(string[i])*(i+1)
    return total

def init_table(n):
    table = []    # declare
    table += [''] * n  # initialize
    return table
    
def hashtable(seq):
    tbl = init_table(len(seq))
    for ele in seq:
        i = hash(ele) % len(seq)

        if tbl[i] == '':
            tbl[i] = ele
        else:  # Separate Chain
            if isinstance(tbl[i], list):
                tbl[i].append(ele)
            else:
                tbl[i] = [tbl[i], ele]
    return tbl

def search(table, item):
    i = hash(item) % len(table)
    if table[i] == '':
        return False
    elif isinstance(table[i], list):
        return item in table[i]
    else:
        return table[i] == item


# Do not modify the following:
data = ['cdab','dbac','dabc','bdac','badc']
data_table = hashtable(data)


```

# Question 3

```python
def hash(string):
    total = 0
    for i in range(len(string)):
        total += ord(string[i])*(i+1)
    return total


def init_table(n):
    table = []    # declare
    table += [''] * n  # initialize
    return table


def hashtable(seq):
    tbl = init_table(len(seq))
    for ele in seq:
        i = hash(ele) % len(tbl)
        if tbl[i] == '':
            tbl[i] = ele
        else:  # Linear probing
            while tbl[i] != '':
                i = (i + 1) % len(tbl)
            tbl[i] = ele
    return tbl

def search(table, item):
    i = hash(item) % len(table)
    checked = 0
    while checked < len(table):
        if table[i] == item:
            return True
        if table[i] == '':
            return False
        i = (i + 1) % len(table)
        checked += 1
    return False


# Do not modify the following:
data = ['cdab','dbac','dabc','bdac','badc']
data_table = hashtable(data)
```
