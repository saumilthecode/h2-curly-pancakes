```python
# TASK 3.1

def hash(string):
    total = 0
    for i in range(len(string)):
        total += ord(string[i])
    return total % 13

# TASK 3.2

def hash_table(seq):
    table = [''] * 13
    for item in seq:
        index = hash(item)
        if table[index] == '':
            table[index] = item
        else:
            if table[index] == item:
                print('item already exists')
            else:
                print(f'collision for {item}')
    return table


keys1 = ('onion','tomato','cabbage', 'okra','banana', 'salt','cucumber','mushroom', 'orange')
table1 = hash_table(keys1)
```
