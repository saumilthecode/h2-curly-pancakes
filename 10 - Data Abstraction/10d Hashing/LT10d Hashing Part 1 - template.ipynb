# LT10d Hashing - Part 1: About Hash Value or Checksum

```python
#Q1.

def hash(string):
    total = 0 
    for i in range(len(string)):
        total += ord(string[i]) *(i+1)
    return total


hash('abcde') % 10
```

```python
#Q2 and 3.

def transmit(data):
    total = 0 
    for i in range(len(data)):
        total += int(data[i]) *((2*i)+1)
    return str(data)+str(total% 10)

    
transmit('12345')
```

```python
#Q4. NRIC

def last_letter(nric):
    total = 0
    letters = ['J', 'Z', 'I', 'H', 'G', 'F','E', 'D', 'C', 'B', 'A']
    multipliers = [2, 7, 6, 5, 4, 3, 2]
    if nric[0] == 'T':
        total += 4
    for i in range(7):
        total += int(nric[i + 1]) * multipliers[i]
    return letters[total % 11]


print(last_letter('S1234567'))
print(last_letter('T1234567'))
```

```python
#Q5. ISBN-10

def isbn(string):
    total = 0
    multipliers = [10, 9, 8, 7, 6, 5, 4, 3, 2]
    for i in range(9):
        total += int(string[i]) * multipliers[i]
    check_digit = (11 - total % 11) % 11
    if check_digit == 10:
        check_digit = 'X'
    else:
        check_digit = str(check_digit)
    return string + check_digit

    
print(isbn('075154926') == '0751549266')
print(isbn('184146208') == '184146208X')
print(isbn('034085045') == '0340850450')
```
