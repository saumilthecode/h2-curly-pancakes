```python
# Copy and Paste the The Queue ADT here:
# Note: Exclude is_empty()

#constructors
def make_empty_queue():
    return []

def make_queue(seq):
    return list(seq)

#modifiers/setters
def enqueue(queue, item):
    queue.append(item)

def dequeue(queue):
    if is_empty(queue):
        return None
    return queue.pop(0)

def clear(queue):
    queue.clear()

#assessors/getters
def front(queue):
    if is_empty(queue):
        return None
    return queue[0]
    
def size(queue):
    return len(queue)
```

```python
# Q9 Print Queue ADT (Extension)

# Constructor
def make_print_queue():
    return [make_empty_queue(), 100]


# Modifiers
def send_job(printq, job):
    enqueue(printq[0], job)


def print_job(printq):
    if size(printq[0]) == 0:
        return None
    if printq[1] == 0:
        print("Please replace the empty ink cartridge.")
        return False
    printed_job = dequeue(printq[0])
    printq[1] -= 1
    return printed_job


def cancel_job(printq, job):
    if job in printq:
        printq.remove(job)
    else:
        return None



def clear_all(printq):
    clear(printq[0])


def replace_cart(printq):
    printq[1] = 100


# Accessors
def next_job(printq):
    return front(printq[0])


def num_jobs(printq):
    return size(printq[0])


def is_pq_empty(printq):
    return size(printq[0]) == 0
```

```python
#Q10  Playlist

def current_song(playlist, minutes):
    # 1. Write code to construct a new queue with songs from playlist
    song_queue = make_queue(playlist)
    if size(song_queue) == 0:
        return None
    songs_completed = int(minutes // 4)
    # 2. Using this queue, use dequeue and emqueue until
    #    the correct song is playing at the front of the queue
    rotations = songs_completed % size(song_queue)
    # 3. Return the song
    for i in range(rotations):
        finished_song = dequeue(song_queue)
        enqueue(song_queue, finished_song)
    return front(song_queue)


# Do not remove/modify the following:
jay_chou = ('silent', 'rainbow', 'nocturne', 'excuse')

# DO NOT run the following CONSECUTIVELY !!!
# Run only one test case each time.

#current_song(jay_chou, 3)   # Expected: silent
#current_song(jay_chou, 4)   # Expected: rainbow
#current_song(jay_chou, 4.1) # Expected: rainbow
#current_song(jay_chou, 7.9) # Expected: rainbow
#current_song(jay_chou, 22)  # Expected: rainbow
#current_song(jay_chou, 45)  # Expected: excuse
#current_song(jay_chou, 90)  # Expected: nocturne
```

```python
#Q11 Pass-the-Bomb

def who_wins(tup, n):
    manq = make_queue(tup)
    while size(manq) != 1:
        for i in range(n):
            player = dequeue(manq)
            enqueue(manq, player)
        dequeue(manq)
    return front(manq)

# Do not modify the following:
players1 = make_queue(('Player1', 'Player2', 'Player3', 'Player4', 'Player5', 'Player6', 'Player7', 'Player8', 'Player9', 'Player10'))
players2 = make_queue(('Player1', 'Player2', 'Player3', 'Player4', 'Player5', 'Player6', 'Player7', 'Player8', 'Player9', 'Player10'))
players3 = make_queue(('Player1', 'Player2', 'Player3', 'Player4', 'Player5', 'Player6', 'Player7', 'Player8', 'Player9', 'Player10'))
who_wins(players1, 2)
```
