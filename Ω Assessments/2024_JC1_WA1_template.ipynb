# Task 2 Count Vowels (3 marks)

```python
#Task 2

def vowels(msg):
    total = 0
    for i in range(len(msg)):
        if msg[i] == 'a' or msg[i] == 'e' or msg[i] == 'i' or msg[i] == 'o' or msg[i] == 'u':
            total +=1
        else:
            pass
    return total



vowels('obama')
```

# Task 3 GPA (6 marks)

```python
#Task 3.1

def GPA(grade):
    if grade == 'A+' or grade == 'A':
        return 5.00
    elif grade == 'A-':
        return 4.50
    elif grade == 'B+':
        return 4.00
    elif grade == 'B':
        return 3.50
    elif grade == 'B-':
        return 3.00
    elif grade == 'C+':
        return 2.50
    elif grade == 'C':
        return 2.00
    elif grade == 'D+':
        return 1.50
    elif grade == 'D':
        return 1.00
    elif grade == 'F':
        return 0.00

GPA('A-')

```

```python
#Task 3.2

def average(grades):
    sum = 0
    for grade in grades:
        sum += GPA(grade)
    return round(sum / len(grades),2)
        
        





#Do not modify the following:
result1 = ('A', 'B-', 'C+')
result2 = ('A', 'B-', 'C+', 'C')

average(result2)
```

# Task 5 Bank (7marks)

```python
#Task 5.1

def FinalAmt():
    ears = int(input('How long do you wish to hold this account '))
    moni = int(input('How much do you wish to deposit '))
    intr = 0
    if ears >= 3:
        print('The interest rate is 3.0% p.a.')
        intr = 1.03
    else:
        print('The interest rate is 2.0% p.a.')
        intr = 1.02
    sum = round( moni * (intr ** ears),2)
    print('The final amount at the end of '+ str(ears) +'years will be: $' + str(sum))
    return sum


FinalAmt()
```
