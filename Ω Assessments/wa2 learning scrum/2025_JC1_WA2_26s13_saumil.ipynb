```python
def remove_digits(text):
    if text == '':
        return text
    else:
        if text[0].isdigit():
            return remove_digits(text[1:])
        else:
            return text[0] + remove_digits(text[1:])
remove_digits("A1B2C3")
```

# Question 3 - Stack ADT [4 marks]

```python
def create_stk(): 
    return []

def push(stk, ele):
    return stk.append(ele)
                           
def pop(stk):
    if len(stk) == 0:
        return None
    return stk.pop()

def is_empty(stk):
    return len(stk) == 0
```

# Question 4 - Application of Stack [6 marks]

```python
def separate(string):
    digit = create_stk()
    alphabetinc = create_stk()
    finalstirng = ''
    while string:
        front = string[0]
        if front.isdigit():
            push(digit,front)
            string = string[1:]
        else:
            push(alphabetinc,front)
            string = string[1:]
    for i in digit:
        string += i
    for i in alphabetinc:
        string += i
    return string

# Write program code for the function separate(string) to separate the digits and alphabet in the string. Your program code will return a new string containing all the digits followed by all the letters in the same order as the original given string.

# For example:

print(separate('62J9M1AG61'))
```

```python
from queue_module import *

# During the registration at a clinic, the senior patients, 60 years old and above, will be placed in a priority queue and the remaining patients in a normal queue.
# The following patients in Group1 are the first to arrive at the clinic:
#     Group1 = (("Sophia", 28), ("James", 45), ("Evelyn", 62), ("Liam", 19), 
#              ("Margaret", 75), ("Noah", 33), ("Dorothy", 68), ("Elijah", 50), 
#              ("Harold", 81), ("Ava", 24))

# Each tuple in the sequence Group1 contains the name and age of the patient.
# Write program code for the function register(group) to allocate the patients to the two queues. 
# After completing the registration, display an output statement to show the number of patients registered:
#              'All the ___ patients have been registered.' 

# You may use the following methods provided in the queue_module.py for the Queue Abstract Data Type (ADT) :
# q.size(): to return the no. of people in the queue, q
# q.enqueue(person): to add a new person into the queue, q. 
# Note that person is a tuple (name,age) containing the person's name and age.
# q.dequeue(): to remove the first person in the queue, q and to return the name of the removed person.
# q.display(): to display the names of all the people in the queue, q.

normal_queue = Queue()
priority_queue = Queue()

def register(group):
    for person in group:
        if int(person[1]) > 60:
            priority_queue.enqueue(person)
        else:
            normal_queue.enqueue(person)
    print('All the ' +str(len(group))+ ' patients have been registered')







#Do not modify the following:
Group1 = (("Sophia", 28), ("James", 45), ("Evelyn", 62), ("Liam", 19), ("Margaret", 75),
          ("Noah", 33), ("Dorothy", 68), ("Elijah", 50), ("Harold", 81), ("Ava", 24))
  
register(Group1)
```
