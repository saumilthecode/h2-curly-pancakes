## Task 1.1 [4 marks]

- the fourth and seventh characters are hyphens
- the first two characters are digits
- the third character is an uppercase letter

# Task 4: Password Verification [8 marks]

A password reset system asks a user to enter a new password and then confirm it.

## Task 4.1 [4 marks]

Write a function `valid_password(password)` that returns `True` if the password is valid, or `False` otherwise.

A valid password must:
- have at least 8 characters
- contain at least one uppercase letter
- contain at least one lowercase letter
- contain at least one digit

````python
def valid_password(password):
    if len(password) < 8:
        return False

    uppr = False
    lowr = False
    numbr = False

    for char in password:
        if char.isupper():
            uppr = True
        elif char.islower():
            lowr = True
        elif char.isdigit():
            numbr = True

    return uppr and lowr and numbr
````

````python
print(valid_password("Abcdef12"))   # True
print(valid_password("abcdef12"))   # False (no uppercase)
print(valid_password("Abc123"))     # False (too short)
print(valid_password("ABCDEFGH"))   # False (no lowercase / no digit)
````

- returns `"Password reset successfully"` if the passwords match
- returns `"Password reset failed"` if all 3 attempts do not match

````python
def valid_password(password):
    if len(password) < 8:
        return False

    uppr = False
    lowr = False
    numbr = False

    for char in password:
        if char.isupper():
            uppr = True
        elif char.islower():
            lowr = True
        elif char.isdigit():
            numbr = True

    return uppr and lowr and numbr

def verify_password():
    password = input("Enter a new password: ")
    if not valid_password(password):
        return "Invalid password"
    attempts = 0
    while attempts < 3:
        reinput = input("Enter the password again for verification: ")
        if password == reinput:
            return "Password reset successfully"
        attempts += 1
    return "Password reset failed"
````

# Task 6: Library Loan Codes [10 marks]

A library loan code is stored as a string in the following format:

    <book code>/<member code>/<days>

For example:

    BK204/M1025/14

- The book code starts with `"BK"` followed by exactly 3 digits.
- The member code starts with `"M"` followed by exactly 4 digits.
- The number of days must be an integer from 1 to 21 inclusive.

## Task 6.1 [5 marks]

Write a function `valid_loan_code(code)` that returns `True` if `code` is valid, or `False` otherwise.

````python
def valid_loan_code(code):
    parts = code.split("/")

    if len(parts) != 3:
        return False

    book_code = parts[0]
    member_code = parts[1]
    days = parts[2]

    if len(book_code) != 5:
        return False

    if book_code[:2] != "BK":
        return False

    if not book_code[2:].isdigit():
        return False

    if len(member_code) != 5:
        return False

    if member_code[0] != "M":
        return False

    if not member_code[1:].isdigit():
        return False

    if not days.isdigit():
        return False

    days = int(days)

    if days < 1 or days > 21:
        return False

    return True
````

## Task 8.2 [3 marks]

Write a function `censor_word(message, banned)` that replaces every occurrence of `banned` in `message` with `"***"`.

The matching is case-sensitive.

For example:

    censor_word("Do not spam the chat with spam", "spam")

returns:

    "Do not *** the chat with ***" 

````python
def censor_word(message, banned):
    words = message.split(" ")
    new_message = ""

    for word in words:
        if word == banned:
            new_message = new_message + "***"
        else:
            new_message = new_message + word

        new_message = new_message + " "

    return new_message[:-1]
````

## Task 8.3 [3 marks]

For example:

    word_frequency("go go study then go rest")

returns:

    {"go": 3, "study": 1, "then": 1, "rest": 1}

````python
def word_frequency(message):
    words = message.split(" ")
    frequency = {}

    for word in words:
        if word in frequency:
            frequency[word] = frequency[word] + 1
        else:
            frequency[word] = 1

    return frequency
````

````python
print(word_frequency("go go study then go rest"))
# {'go': 3, 'study': 1, 'then': 1, 'rest': 1}
````

# Task 10: Event Equipment Loan Tracking [11 marks]

An event coordinator tracks items loaned out to different venues. Each transaction is given a unique reference code structured in the following format:
`<venue_code>#<item_id>#<quantity>`

For example:

`HALL1#EQ104#12`

`LAB02#EQ915#05`

- Venue Code: A 5-character alphanumeric string that must be entirely in uppercase letters and numbers (e.g., HALL1, LAB02).

- Item ID: Starts with the uppercase string "EQ" followed by exactly 3 digits (e.g., EQ104).

- Quantity: A 2-digit numeric string ranging from "01" to "50" inclusive.

## Task 10.1 [5 marks]

Write a function `valid_loan_code(code)` that takes a string `code` and returns `True` if it strictly matches the formatting criteria above, or `False` otherwise.

````python
def valid_loan_code(code):
    parts = code.split("#")

    if len(parts) != 3:
        return False

    venue_code = parts[0]
    item_id = parts[1]
    quantity = parts[2]

    if len(venue_code) != 5:
        return False

    if not venue_code.isalnum():
        return False

    if venue_code != venue_code.upper():
        return False

    if len(item_id) != 5:
        return False

    if item_id[0:2] != "EQ":
        return False

    if not item_id[2:5].isdigit():
        return False

    if len(quantity) != 2:
        return False

    if not quantity.isdigit():
        return False

    if int(quantity) < 1 or int(quantity) > 50:
        return False

    return True
````

````python
print(valid_loan_code("HALL1#EQ104#12"))  # True
print(valid_loan_code("LAB02#EQ915#05"))  # True
print(valid_loan_code("hall1#EQ104#12"))  # False (venue code not uppercase)
print(valid_loan_code("HALL1#EQ10#12"))   # False (item ID wrong length)
print(valid_loan_code("HALL1#EQ104#55"))  # False (quantity out of range)
````

## Task 10.2 [3 marks]

Write a function `aggregate_equipment(loans)` that takes a list of valid loan codes and returns a dictionary. The dictionary keys must be the unique `item_id` values, and the associated values must be the total accumulated integer quantity requested for that item across all records.

````python
def aggregate_equipment(loans):
    equipment = {}

    for loan in loans:
        parts = loan.split("#")
        item_id = parts[1]
        quantity = int(parts[2])

        if item_id in equipment:
            equipment[item_id] = equipment[item_id] + quantity
        else:
            equipment[item_id] = quantity

    return equipment
    
# Do not modify the following:
sample_loans = ["HALL1#EQ104#12", "LAB02#EQ915#05", "AUDI3#EQ104#08"]

print(aggregate_equipment(sample_loans))        # Output: {'EQ104': 20, 'EQ915': 5}
````
