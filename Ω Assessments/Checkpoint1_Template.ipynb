# Question 1 : Estimate height from age

Write a code to do the following:

* prompt the user to enter a child's age in years
* use the formula given below to calculate the height in centimetres
* print an output indicating the estimated height in centimetres 



```python
# Write your program code below:
#height (cm) = age (yr) x 6.5 + 76 (cm)

age = int(input("Please enter your age in years: "))

height = (age * 6.5) + 76



print("your height should be " + str(height))






```

# Question 2 : Convert a phone number format.

Write a code to do the following:

* prompt the user to enter an 8 digit phone number without spacing
* use string slicing and concatenation to convert the number to "+65 XXXX XXXX" format
* print an output showing the converted number format

```python
# Write your program code below:

number = input("Please enter your phone number without the country code or spacing: ")
if len(number) == 8:
    first_4 = number[:4]
    last_4 = number[4:]
    space = ' '
    print('+65'+ space + first_4 + space + last_4)
else:
    print("invalid")




```

# Question 3 : Simple Calculator

Write code to do the following:

- prompt the user to enter 2 numbers, n1 and n2, and an operator character char which is a string '+', '-', '*' or '/'.
- execute the chosen math operation based on the input operator char according to the following cases:
    - '+' : add both n1 and n2
    - '-' : subtract n2 from n1
    - '*' : multiply n1 and n2
    - '/' : divide n1 by n2; check n2 to avoid division by zero error.
- print out the respective answer, or output a message "Operation Not Executed" if the operator or answer cannot be found.

```python
# Write your program code below:
#"simple" calculator therefore no input checking
n1 = input("Please input your first number: ")
n2 = input("Please input your second number: ")
char = input("Please input your operator (+,-,*,/): ")

n1 = int(n1)
n2 = int(n2)
if char == '/':
    if n2 == 0 :
        print("Operation Not Executed")
    else:
        print(n1/n2)
elif char == '+':
    print(n1+n2)
elif char == '-':
    print(n1-n2)
elif char == '*':
    print(n1*n2)
else:
    print("Operation Not Executed")
```
