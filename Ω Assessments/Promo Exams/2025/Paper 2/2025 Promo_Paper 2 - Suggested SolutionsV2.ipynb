```python
# Task 5.1 [5m]

lst = [5, 3, 10, 7, 4, 9, 2, 8, 6, 7]     # This list is provided for you

def Partition(Seq, Start, End):

    PivotValue = Seq[Start]
    Lo = Start + 1
    Hi = End                                     # 1m Correct use of
    
    while Lo <= Hi:                              #    Pseudocode
        while Lo <= Hi and Seq[Lo] < PivotValue: #    Indentation
            Lo += 1                              #    Var names/logic
            
        while Lo <= Hi and Seq[Hi] >= PivotValue:
            Hi -= 1                              # 1m fill in Hi - 1
            
        if Lo < Hi:
            Seq[Lo], Seq[Hi] = Seq[Hi], Seq[Lo]  # 1m Code to swap

    Seq[Hi], Seq[Start] = Seq[Start], Seq[Hi]    # 1m Code to swap
    
    return Hi
```

```python
# Task 6 [4m]

lstnum = [2, 5, 12, 20, 35, 70, 99, 105, 225, 364, 600, 930, 1222, 2002, 2502, 3010, 4666]

def binarysearch(lst, item):
    start = 0
    end = len(lst) - 1
    while start <= end:            # 1m Iterate
        mid = (start + end) // 2   # 1m check value at median position
        if lst[mid] == item:
            return True
        elif lst[mid] < item:
            start = mid + 1        # 1m update pointers
        else:
            end = mid - 1
    return False                   # 1m correct return
```

```python
# Task 7.1 [2m]

def hashfn(string):
    total = 0
    for i in range(len(string)):
        total += ord(string[i]) * 17 ** (i + 1)   # 1m correct calculation 
    return total % 25                             # 1m correct iteration and return
```

```python
# Task 7.2 [6m]

names = ["Snuvla", "Tordenk", "Blommig", "Kappar", "Fastig", "Lovkant",
    "Dromtorp", "Kullstad", "Vanklig", "Mornbak", "Trolka", "Gursnad", "Norby", "Piltra",
    "Somnig", "Frastorp", "Dolmig", "Halska", "Skurna", "Valkom", "Krundal", "Ornsta",
    "Flaktig", "Jornka", "Grynskar", "Stubbrod", "Barnig", "Zalvar", "Tallhov", "Holmbrak"]

def hashtable(lst):
    hashtable = [''] * 25
    
    for ele in lst:                        # 1m initialise and iterate
        hsh = hashfn(ele)                  # 1m compute hash
        if hashtable[hsh] == '':           # 1m empty case
            hashtable[hsh] = ele
        elif type(hashtable[hsh]) == list: # 1m list case
            hashtable[hsh].append(ele)
        else:                              # 1m non-empty string case 
            hashtable[hsh] = [hashtable[hsh], ele]
    return hashtable
```

```python
hashtable(names)                           # 1m return and display output
```
