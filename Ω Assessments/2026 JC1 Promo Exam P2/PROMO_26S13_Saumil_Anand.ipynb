## Task 2.1 — Read Scores [4 marks]

```python
def read_scores(filename):
    lines = ''
    lost = []
    with open(filename) as f:
        lines = f.readlines()
        f.close()
    for line in lines[1:]:
        line = line.replace('\n', '')
        CompID, Name, JC, S1, S2, S3, S4, S5, S6, S7 = line.strip(' ').split(',')
        lost.append((CompID, float(S1), float(S2), float(S3), float(S4), float(S5), float(S6), float(S7)))
    return lost
```

## Task 2.2 — Calculate Results [3 marks]

```python
def calculate_results(competitors):
    lost = []
    for i in range(len(competitors)):
        CompID, S1, S2, S3, S4, S5, S6, S7 = competitors[i]
        lst = [S1, S2, S3, S4, S5, S6, S7]
        lowest = min(lst)
        largest = max(lst)
        total = 0
        for i in range(len(lst)):
            total += lst[i]
        total = total - lowest - largest
        lost.append((CompID, round(total / 5, 2)))
    return lost
```

## Task 3 — Collatz Peak Value [5 marks]

```python
def collatz(n):
    if type(n) != list:
        lst = [n,]
    else:
        lst = n
    if lst[-1] == 1:
        return max(lst)
    elif lst[-1] % 2 == 0:
        number_generated = lst[-1] // 2
        lst.append(number_generated)
        return collatz(lst)
    elif lst[-1] % 2 == 1 and lst[-1] > 1:
        number_generated = 3 * lst[-1] + 1
        lst.append(number_generated)
        return collatz(lst)
```

## Task 5 — Bubble Sort [4 marks]


```python
fleet = [
    ("Blue Horizon", 95),
    ("Cloud Nine", 156),
    ("Dream Chaser", 141),
    ("Good Vibes", 95),
    ("High Flyer", 72),
    ("Sky Dancer", 128),
    ("Sunny Side Up", 84),
    ("Wanderlust", 103)
]

#############################
# Do not edit above this line
#############################

def bubble_sort(lst):
    for z in range(len(lst)):
        swapped = False
        for i in range(len(lst) - z - 1):
            if lst[i][1] > lst[i + 1][1]:
                lst[i], lst[i + 1] = lst[i + 1], lst[i]
                swapped = True
        if not swapped:
            break
```

## Task 6.1 — Implement `merge` [2 marks]


```python
def merge(left, right):
    merged = []
    while left and right:
        if left[0] > right[0]:
            merged.append(left[0])
            left = left[1:]
        else:
            merged.append(right[0])
            right = right[1:]
    merged += left
    merged += right
    return merged
```

## Task 6.2 — Complete the pseudocode [3 marks]

## Task 6.3 — Implement `merge_sort` [2 marks]


```python
def merge_sort(seq):
    if len(seq) <= 1:
        return seq
    mid = len(seq) // 2
    left = merge_sort(seq[:mid])
    right = merge_sort(seq[mid:])
    return merge(left, right)
```

## Task 7.1 — Hash Function [3 marks]


```python
def hash_name(item):
    total = 0
    counter = 1
    for i in item:
        total += ord(i)*counter
        counter += 1
    total = total % 25
    return total
```

## Test cases for Task 7.1

## Task 7.2 — Populate Hash Table [3 marks]


```python
team_table = ['']*25

for i in range(len(teams)):
    index = hash_name(teams[i][0])
    if team_table[index] == '':
        team_table[index] = teams[i]
    elif type(team_table[index]) == tuple:
        team_table[index] = [team_table[index], teams[i]]
    else:
        team_table[index].append(teams[i])
```

## Task 7.3 — Search Hash Table [2 marks]

```python
def find_goals(table, team_name):
    hh = hash_name(team_name)
    if table[hh] != '':
        if type(table[hh]) == list:
            for i in range(len(table[hh])):
                if table[hh][i][0] == team_name:
                    return table[hh][i][1]
        else:
            if table[hh][0] == team_name:
                return table[hh][1]
    return None
```

## Task 8.1 — Insert into a BST [3 marks]

```python
from module import *

team_codes = [
    "LIV", "OLD", "CHE", "FUL", "MNV",
    "BRE", "WOL", "EVE", "MNT", "BOU",
    "WHU", "LEE", "NOT", "BUR", "LON",
    "CRY", "SUN", "BRI", "WOO", "AST"
]

#############################
# Do not edit above this line
#############################

def insert_item(tree, item):
    if is_empty(tree):
        return make_tree(item, make_empty_tree(), make_empty_tree())
    if item > entry(tree):
        return make_tree(entry(tree), left_branch(tree),
                         insert_item(right_branch(tree), item))
    else:
        return make_tree(entry(tree), insert_item(left_branch(tree), item),
                         right_branch(tree))

```

## Task 8.2 — Search a BST [3 marks]

```python
def search_tree(tree, target):
    if is_empty(tree):
        return False
    if target == entry(tree):
        return True
    if target > entry(tree):
        return search_tree(right_branch(tree), target)
    else:
        return search_tree(left_branch(tree), target)
```

## Task 8.3 — In-order traversal [2 marks]

```python
def traversal(tree):
    if is_empty(tree):
        return []
    return traversal(left_branch(tree)) + [entry(tree)] + traversal(right_branch(tree))
```

## Task 9.1 — Validate Hashtag [3 marks]


```python
def is_valid(tag):
    if tag != '' and tag[0] == '#':
        tag = tag[1:]
        if tag.isalnum():
            if 2 <= len(tag) <= 14:
                return True
    return False
```

## Task 9.2 — Count Hashtags [4 marks]

```python
def compile_hashtags(lst):
    dic = {}
    for i in range(len(lst)):
        if not is_valid(lst[i]):
            pass
        else:
            hashtag = lst[i]
            hashta = hashtag.lower()
            if hashta not in dic:
                dic[hashta] = 1
            else:
                dic[hashta] += 1
    return dic
```
