#### Task 6.1

```python
#Task 6.1

def hash(num):
    sam = 0
    for i in range(len(num)):
        sam += ord(num[i]) * (len(num)-i)
    return sam % 100


hash('SBK4125T')
```

```python
# Test case for Task 6.1

print(hash('SBK4125T') == 72)
```

#### Task 6.2

```python
#Task 6.2

data_table = [''] * 100
with open('VEHICLES.txt') as f:
    lines = f.readlines()
    f.close()
for line in lines[1:]:
    Vehicle_number,Workshop,Date,Status = line.strip().split(',')
    hashval = hash(Vehicle_number)
    if data_table[hashval] == '':
        data_table[hashval] = (Vehicle_number,Workshop,Date,Status)
    else:
        print('collision')

data_table
    

```

#### Task 6.3

```python
#Task 6.3

def query(num):
    hashval = hash(num)
    if data_table[hashval][0]== num:
        return data_table[hashval][3]
    else:
        print('car not found')
query('STU3456T')
```

```python
# Test case for Task 6.3

print(query('STU3456T') == 'Pending for repair')
```

#### Task 6.4

```python
#Task 6.4

def ready(num):
    hashval = hash(num)
    if data_table[hashval][0]== num:
        data_table[hashval][3] == 'ready for collection'
        print('Data updated')
    else:
        print('vehickle not found')
```

```python
# Test case for Task 6.4

ready('XYZ9012Y')
print(query('XYZ9012Y'))
```
