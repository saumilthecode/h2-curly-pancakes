```python
#Task 1.1

def discriminant(a, b, c):
    discri = b**2 - 4 * a * c   # [1m] correct calculation
    if discri < 0:
        return 'No solution'            
    else:
        return discri           # [1m] correct if-else check and return
    
print(discriminant(2, 1, 2) == 'No solution')
print(discriminant(2, 3, -2) == 25)   
print(discriminant(2, 4, 2) == 0) # [1m] 3 correct and relevant test cases (1 positive, 1 negative, 1 zero)
```

```python
#Task 3.1

ridership_list = [] 
file = open('RIDERSHIP.csv')        # given

next(file)                          # [1m] skip header row
for line in file:                   # [1m] iterate thru file
    line = line.strip()             # [1m] strip      
    line = line.split(',')           # [1m] split
    ridership_list.append(line)     # [1m] append
    
    
file.close()                        # given
ridership_list
```

```python
#Task 3.2

def total_ridership(lst, tpt):
    total = 0                       # [1m] init and return total
    for line in lst:
        if line[1] == tpt:          # [1m] check for matching tpt
            total += int(line[2])   # [1m] correct int conversion + add

    return total

total_ridership(ridership_list, 'MRT')
```

```python
#Task 3.3

def avg_ridership(lst, tpt):
    years = 0
    for line in lst:
        if line[1] == tpt and int(line[2]) > 0: # [1m] check for correct tpt, years += 1
            years += 1                          
    return total_ridership(lst, tpt) / years    # [1m] helper function used

avg_ridership(ridership_list, 'Taxi')
```

```python
# Task 5.2

def MergeSort(seq):              # 1m for correct merge sort
    if len(seq) <= 1:         
        return seq
    # Divide the list into two halves
    mid = len(seq) // 2
    left = seq[:mid]          
    right = seq[mid:]        
    # Recursively sort both halves
    left = MergeSort(left)          
    right = MergeSort(right)           
    # Merge the sorted halves
    return Merge(left, right)

def Merge(left, right):               # 1m for correct merge
    result = []
    while left and right:                      
        if left[0] <= right[0]:                        
            result.append(left.pop(0))         
        else:
            result.append(right.pop(0))        
    if len(left) == 0:
        result += right   
    else:
        result += left   

    return result                       # 1m for using pseudo-code structure


MergeSort([
11.5, 15.8, 10.3, 13.0, 12.1, 9.5, 14.8, 
11.2, 10.0, 13.7, 11.0, 12.4, 17.6, 12.0, 
12.5, 11.8, 9.9, 10.2, 14.3, 12.3
]
)
```

```python
# Task 5.3

def sorting(data):       # 1m for using task 5.2 algo, or any other sorting algo
    if len(data) <= 1:
        return data

    mid = len(data) // 2
    left = data[:mid]
    right = data[mid:]

    left = sorting(left)
    right = sorting(right)

    return merging(left, right)

def merging(left, right):
    result = []

    while left and right:
        if float(left[0][1]) <= float(right[0][1]):
            result.append(left.pop(0))    # 1m for tuple indexing, 1m for float
        else:                                     
            result.append(right.pop(0))

    if len(left) == 0:
        result += right
    else:
        result += left   

    return result


sorting([ 
('ELAINE', '12.5'), ('GEMMA', '14.3'), ('JACK', '12.0'), ('OLIVE', '11.5'), ('QUINN', '10.2'), ('CASSIE', '9.5'), ('ALEX', '13.7'), ('PETER', '12.1'), ('TOM', '11.2'), ('MIA', '12.3'), ('ROSE', '14.8'), ('BEN', '11.0'), 
('DAN', '12.4'), ('FRED', '9.9'), ('LEE', '17.6'), 
('HARRY', '11.8'), ('IRIS', '10.0'), ('KELLY', '15.8'), ('NICK', '10.3'), ('SARA', '13.0')
]
)
```

```python
#Task 6.1 (3 marks)

def hash(num):
    total = 0
    i = 8
    for ele in num:             #[1] iteration
        total += ord(ele) * i   #[1] multiple with i
        i -= 1
    return total % 100          #[1] total and %

hash('SBK4125T')
        
```

```python
#Task 6.2 (6 marks)

data_table = ['']*100            #[1]

file = open('VEHICLES.txt')

lines = file.readlines()         #[1]open,read,close file

for line in lines[1:]:
    line = line.strip()
    [veh_num, workshop, date, status] = line.split(',') #[1]strip and split
    
    hash_value = hash(veh_num)      #[1] compute hash value
    if data_table[hash_value] == '':  #[1] check empty
        data_table[hash_value] = [veh_num, workshop, date, status] #[1] assign data
    else:
        print(veh_num, 'not stored.')

file.close()

print(data_table)

```

```python
#Task 6.3 (5 marks)

def query(num):
    hash_value = hash(num)                 #[1] compute hash value
    if data_table[hash_value] == '':       #[1] check empty ''
        return 'Vehicle not found.'
    if num == data_table[hash_value][0]:   #[1] check vehicle num
        return data_table[hash_value][3]   #[1] return status
    else:
        return 'Vehicle not found.'        #[1] Vehicle not found
        
print(query('STU3456T'))
print(query('ABC9876D'))            #check empty ''
print(query('XYZ3456D'))            #another veh num with same hash value
```
