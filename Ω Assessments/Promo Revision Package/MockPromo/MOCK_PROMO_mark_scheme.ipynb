## Question 1: Equipment Inspection — Suggested Solution [6 marks]

## Task 1.1 — Validated inspection input [4 marks]


```python
def collect_inspection(input_function=input):
    inspector = input_function("Inspector name: ")
    while inspector == "" or any(character.isdigit() for character in inspector):
        inspector = input_function("Invalid name. Inspector name: ")  # [1m]

    number = int(input_function("Number of items: "))
    while number < 1 or number > 6:
        number = int(input_function("Invalid number. Number of items: "))  # [1m]

    readings = []
    while len(readings) < number:                              # [1m]
        reading = int(input_function("Vibration reading: "))
        while reading < 0 or reading > 100:
            reading = int(input_function("Invalid reading. Vibration reading: "))
        readings.append(reading)                               # [1m]

    return inspector, readings


test_inputs = iter(["Mira Tan", "5", "42", "75", "68", "91", "54"])
inspector, readings = collect_inspection(lambda prompt: next(test_inputs))
print(inspector, readings)
```

## Task 1.2 — Inspection summary [2 marks]


```python
average = round(sum(readings) / len(readings), 1)  # [1m]
attention = 0
for reading in readings:
    if reading >= 70:
        attention += 1                                # [1m]

print("Inspector:", inspector)
print("Readings:", readings)
print("Average:", average)
print("Items requiring attention:", attention)
```

## Question 2: Rainfall Stations — Suggested Solution [7 marks]

## Task 2.1 — Read rainfall data [4 marks]


```python
def read_rainfall(filename):
    stations = []
    with open(filename) as file:
        lines = file.readlines()                       # [1m]

    for line in lines[1:]:                             # [1m]
        fields = line.strip().split(",")
        readings = [float(value) for value in fields[2:]]  # [1m]
        stations.append((fields[0], *readings))        # [1m]
    return stations


stations = read_rainfall("RAINFALL.txt")
print(stations[:2])
```

## Task 2.2 — Summarise rainfall [3 marks]


```python
def summarise_rainfall(stations):
    summaries = []
    for station in stations:                           # [1m]
        readings = station[1:]
        total = round(sum(readings), 1)                # [1m]
        spread = round(max(readings) - min(readings), 1)  # [1m]
        summaries.append((station[0], total, spread))
    return summaries


print(summarise_rainfall(stations)[:2])
```

## Question 3: Container Seal Codes — Suggested Solution [5 marks]

## Task 3 — Recursive character count [5 marks]


```python
def count_character(code, target):
    if code == "":                                     # [1m]
        return 0
    if code[0] == target:                              # [1m]
        current = 1                                    # [1m]
    else:
        current = 0
    remaining = count_character(code[1:], target)      # [1m]
    return current + remaining                         # [1m]


print(count_character("A7B7C77", "7"))
print(count_character("SEAL", "X"))
```

forming the current contribution, recursively processing the remaining string, and

## Question 4: Round-Robin Print Scheduling — Suggested Solution [8 marks]

## Task 4.1 — Process one turn [5 marks]


```python
from module import *

jobs = (("P101", 2), ("P102", 1), ("P103", 3))

print_queue = make_empty_queue()
for job in jobs:
    enqueue(print_queue, job)


def process_turn(queue):
    if is_empty_queue(queue):                         # [1m]
        return None

    job = dequeue(queue)                              # [1m]
    pages_remaining = job[1] - 1                     # [1m]
    if pages_remaining > 0:
        enqueue(queue, (job[0], pages_remaining))     # [1m]
        return None
    return job[0]                                     # [1m]
```

## Task 4.2 — Determine the completion order [3 marks]


```python
def completion_order(queue):
    completed = []                                    # [1m]
    while not is_empty_queue(queue):                  # [1m]
        job_code = process_turn(queue)
        if job_code is not None:
            completed.append(job_code)                # [1m]
    return completed


print(completion_order(print_queue))
```

- Task 4.2: initialise the result list, repeatedly call `process_turn()` until the queue
  is empty, and store only returned job codes in completion order, 1 mark each. The

## Question 5: Bubble Sort — Suggested Solution [4 marks]

## Task 5 — Sort laboratory samples [4 marks]


```python
samples = [
    ("S104", 3), ("S101", 5), ("S106", 2),
    ("S103", 5), ("S105", 1), ("S102", 4)
]


def bubble_sort(samples):
    swaps = 0
    for end in range(len(samples) - 1, 0, -1):         # [1m]
        for index in range(end):
            if samples[index][1] < samples[index + 1][1]:  # [1m]
                samples[index], samples[index + 1] = samples[index + 1], samples[index]  # [1m]
                swaps += 1
    return swaps                                       # [1m]


print(bubble_sort(samples))
print(samples)
```

Award 1 mark each for repeated passes, correct descending adjacent comparison without
swapping equal priorities, swapping complete tuples while counting swaps, and returning

## Question 6: Trail Backtracking — Suggested Solution [8 marks]

## Task 6.1 — Stack operations [3 marks]


```python
from module import *

checkpoints = ("ENTRANCE", "PINE", "RIDGE", "LAKE")


def push(stack, checkpoint):
    stack.append(checkpoint)                           # [1m]
    return stack


def pop(stack):
    if is_empty_stack(stack):                          # [1m]
        return None
    return stack.pop()                                 # [1m]


trail = make_empty_stack()
for checkpoint in checkpoints:
    push(trail, checkpoint)
print(trail)
```

## Task 6.2 — Peek at the stack [2 marks]


```python
def peek(stack):
    if is_empty_stack(stack):                          # [1m]
        return None
    return stack[-1]                                   # [1m]
```

## Task 6.3 — Backtrack [3 marks]


```python
def backtrack(stack):
    route = []                                         # [1m]
    while not is_empty_stack(stack):                   # [1m]
        route.append(pop(stack))
    return route                                       # [1m]


print(peek(trail))
print(backtrack(trail))
```

- Task 6.2: handle empty and return the top item without removal, 1 mark each.
- Task 6.3: initialise a result, repeatedly pop until empty, and return LIFO order, 1 mark each.


## Question 7: Archive Boxes — Suggested Solution [6 marks]


```python
box_codes = [
    "A104", "A118", "A205", "B107", "B215",
    "B412", "C103", "C220", "D114", "D305",
    "E102", "E211", "F109", "F330", "G205"
]
```

## Task 7.1 — Iterative binary search [5 marks]


```python
def find_box(codes, target):
    low = 0
    high = len(codes) - 1
    comparisons = 0                                   # [1m]
    while low <= high:                                 # [1m]
        middle = (low + high) // 2
        comparisons += 1
        if codes[middle] == target:
            return middle, comparisons                 # [1m]
        if codes[middle] < target:
            low = middle + 1                           # [1m]
        else:
            high = middle - 1                          # [1m]
    return -1, comparisons
```

## Task 7.2 — Test the search [1 mark]


```python
print(find_box(box_codes, "B412"))          # [1m] present case
print(find_box(box_codes, "D200"))                    # and absent case
```

## Question 8: Parcel-Locker Hash Table — Suggested Solution [8 marks]


```python
from module import hash_locker

parcels = [
    ("L204", "Amir"), ("L213", "Beatrice"),
    ("L222", "Chen Wei"), ("L315", "Divya"),
    ("L324", "Ethan"), ("L435", "Farah")
]
```

## Task 8.1 — Generate the probe sequence [2 marks]


```python
def probe_sequence(initial_index, table_size):
    sequence = [initial_index]                         # [1m]
    current = (initial_index + 2) % table_size
    while current != initial_index:
        sequence.append(current)                       # [1m]
        current = (current + 2) % table_size
    return sequence


print(probe_sequence(8, 11))
```

## Task 8.2 — Insert parcels [4 marks]


```python
def insert_parcel(table, record):
    initial_index = hash_locker(record[0], len(table))  # [1m]
    for index in probe_sequence(initial_index, len(table)):  # [1m]
        if table[index] is None:
            table[index] = record                       # [1m]
            return True
    return False                                         # [1m]


parcel_table = [None] * 11
for parcel in parcels:
    insert_parcel(parcel_table, parcel)
print(parcel_table)
```

## Task 8.3 — Search for a parcel [2 marks]


```python
def find_parcel(table, locker_code):
    initial_index = hash_locker(locker_code, len(table))
    for index in probe_sequence(initial_index, len(table)):  # [1m]
        if table[index] is None:
            return None
        if table[index][0] == locker_code:
            return table[index]                              # [1m]
    return None


print(find_parcel(parcel_table, "L222"))
print(find_parcel(parcel_table, "L999"))
```

## Question 9: Wildlife Survey — Suggested Solution [8 marks]


```python
sightings = [
    ("Hornbill", 18), ("Otter", 7), ("Macaque", 24),
    ("Kingfisher", 12), ("Monitor Lizard", 7),
    ("Wild Boar", 15), ("Pangolin", 3), ("Civet", 12)
]
```

## Task 9.1 — Partition the records [3 marks]


```python
def partition(records):
    pivot = records[0][1]                              # [1m]
    less = []
    equal = []
    greater = []                                       # [1m]
    for record in records:
        if record[1] < pivot:
            less.append(record)
        elif record[1] == pivot:
            equal.append(record)
        else:
            greater.append(record)                     # [1m]
    return less, equal, greater
```

## Task 9.2 — Out-of-place quicksort [5 marks]


```python
def quick_sort(records):
    if len(records) <= 1:
        return records.copy()                          # [1m]
    less, equal, greater = partition(records)          # [1m]
    sorted_less = quick_sort(less)                     # [1m]
    sorted_greater = quick_sort(greater)               # [1m]
    return sorted_less + equal + sorted_greater        # [1m]


original = sightings.copy()
print(quick_sort(sightings))
print(sightings == original)
```

- Task 9.1: obtain the first record's count as pivot, create three new lists, and distribute every complete record correctly, 1 mark each.
- Task 9.2: base case returning a new list, call partition, recursively sort less, recursively sort greater, and concatenate in ascending order, 1 mark each.
