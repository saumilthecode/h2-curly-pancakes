```python
#Task 3.2

def insertion_sort(lst):   
    for k in range(1,len(lst)):     #1m for loop
        i = k  
        while i > 0 and int(lst[i-1][2]) < int(lst[i][2]) :  #1m i>0  #1m compare temperature
            lst[i-1], lst[i] = lst[i], lst[i-1]   #1m swap
            i -= 1   # 1m while loop and decrement of i
    return lst
```

```python
# Task 4

def sum_of_digits(n):  # (integer div method)
    if n == 0:
        return 0       # 1m base case
    digit = n % 10     # 1m compute a digit
    newn = n // 10     # 1m recursion on remaining
    return digit + sum_of_digits(newn) # 1m combine + correctly

def sum_of_digits(n):  # (slicing method)
    if n < 10:
        return n         # 1m base case
    nstr = str(n)
    digit = int(nstr[0]) # 1m compute a digit

    newn = int(nstr[1:]) # 1m recursion on remaining digits
    return digit + sum_of_digits(newn) # 1m combine + correctly

sum_of_digits(3156)
```

# Question 6 (9 marks)

```python
# Task 6.1

def hash(email):
    index = email.index('@')  # 1m slice correct part of email
    left = email[0:index]
    total = 0
    for char in left:         # 1m loop and total
        total += ord(char)
    return total % 20         # 1m use of ord and % 20

hash('abc@yijc.edu.sg')
```

```python
# Task 6.2

def hashtable(lst):
    tbl = ['']*20                         #1m init empty table
    for email in lst:
        i = hash(email)                   #1m use of hash index
        if tbl[i] == '':                  #1m case 1 if empty
            tbl[i] = email
        else:
            j = (i+1)%20                  #1m case 2 linear probing
            for lin_probe in range(19):   #1m inner loop from i/i+1
                if tbl[j] == '':          #
                    tbl[j] = email
                    break
                else:
                    j = (j+1)%20          #1m wrap around
            if j == i:
                print("Insertion failed, table full")
    return tbl
```
