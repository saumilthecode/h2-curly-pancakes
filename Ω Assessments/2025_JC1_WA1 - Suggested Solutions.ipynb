# Q1 Luggage Tag (4 marks)

```python
#Q1

def generate(first_name, last_name, booking_code):
    part1 = last_name[:3].upper()     # [1] slice first 2 or 3 char
    part2 = first_name[:2].upper()    # [1] use of .upper
    part3 = booking_code[-4:]          # [1] slice last 4 characters
    return part1 + part2 + part3      # [1] correct concatenation and return

generate("Rachel", "Nguyen", "BX34YZ21") == "NGURAYZ21"
```

# Q2 Postal Code Check (5 marks)

```python
#Q2

postalcode = input("Please enter a postal code:")  # [1] code input with prompt
                               
if len(postalcode) != 6:                           # [1] length check
    print('Invalid input')                       
elif not postalcode.isdigit():                     # [1] format check
    print('Invalid input')
else:
    first2digit = int(postalcode[0:2])             # indexing the first 2 digits
    if 1 <= first2digit <= 15:                     # [1] to do a range check
        print('This is the Central Region.')
    elif 16 <= first2digit <= 41:          
        print('This is the East Region.')
    elif 42 <= first2digit <= 52 or 80 <= first2digit <= 99:          
        print('This is North-East Region.')
    elif 53 <= first2digit <= 57 or 72 <= first2digit <= 78:          
        print('This is North Region.')
    elif 58 <= first2digit <= 71:          
        print('This is West Region.')
    else:
        print('Region not classified.')            # [1] correct if-elif-else logic
```

# Q3 Flowchart (6 marks)

```python
#Q3

def div19(num):
    while num > 19:                    # [1] while loop with condition
        print("Number:", num)
        numstr = str(num)              # [1] correct types (str now int later)
        lastdigit = int(numstr[-1])    # 
        firstfew = int(numstr[:-1])    # [1] slicing can handle flexible length
                                       #     modulo 10 also accepted
        print("Firstfew:", firstfew, "Last:", lastdigit) 
        num = firstfew + lastdigit * 2 # [1] correct number outputs
                                       # [1] compute new num
    if num == 19:                      
        return True
    else:
        return False                   # [1] implement return

    
div19(532) == True
```

# Q4 Area of Triangle (2 marks)

```python
#Q4

def area_triangle(a, b, c):
    s = (a + b + c) / 2                              
    area = (s * (s - a) * (s - b) * (s - c)) ** 0.5  # [1] use of formula
    return round(area, 2)               # [1] round to 2dp and return 


area_triangle(3, 4, 5) == 6
```

# Q5 Classify the Temperature (3 marks)

```python
#Q5

def classify(temp):
    if temp < 0:               #[1] use if-elif-else
        return "Freezing"      #[2] all correct output
    elif temp < 10:            #[-1] each wrong output
        return "Cold"
    elif temp < 20:
        return "Cool"
    elif temp < 30:
        return "Warm"
    else:                      # temp=30 is 'Hot'
        return "Hot"
```

```python
print(classify(-5)=="Freezing")
print(classify(8)=="Cold")
print(classify(15)=="Cool")
print(classify(25)=="Warm")
print(classify(35)=="Hot")
```

# Q6 Average Height (5 marks)

```python
#Q6

#[-1] Not referring to the height in the tuple correctly.
#[-1] Not rounding to 2 dp
#[-1] Wrong way to round to 2 dp

def average(data):
    total = 0                    
    
    for student in data:         #[1] iterate thru data
        height = student[1]      #[1] correctly referring to height
        total += height          #[1] add height to total
        
    average = total/len(data)    #[1] compute average

    return round(average, 2)     #[1] round to 2 dp


#Do not modify the following:
students = (("Alice", 160), ("Bob", 175), ("Charlie", 168), ("Diana", 170), ("Ethan", 165))
students2 = (("Alice", 161.5), ("Bob", 175), ("Diana", 170), ("Ethan", 165), )

print(average(students)==167.6)
print(average(students2)==167.88)
```

# Q7 Modify a Tuple (5 marks)

```python
#Q7

#[-2] Do not use list for a tuple question.
#[-1] Not referring to the name in the tuple correctly.
#[-1] Not returning a tuple of tuples.
#[-1] You have hardcoded students.
#[-2] Tuples are immutable. Cannot replace by assignment.

def modify(data, new):

    tup = ()
    for student in data:           #[1] iterate thru data
        if student[0] == new[0]:   #[1] check name
            tup += (new,)          #[1] replace with new tuple
        else:
            tup += (student,)      #[1] add student tuple

    return tup                     #[1] return tup


#Do not modify the following:
students = (("Alice", 160), ("Bob", 175), ("Charlie", 168), ("Diana", 170), ("Ethan", 165))
charlie = ("Charlie", 170)

modify(students, charlie)
```

# Q8 Encrypt a Secret Message (5 marks)

```python
#Q8

#[-1] Not checking the vowel correctly.
#[-2] Wrong way to replace with *.
#[-1] Not check both lower and upper cases.
#[-1] Wrong syntax to append the character and *

def encrypt(secret):
    
    msg = ''
    
    for char in secret:             #[1] iterate thru the string
        if char in 'aeiouAEIOU':    #[2] check for vowel, lower+upper case
            msg += '*'              #[1] replace with *
        else:
            msg += char             #[1] append char
    
    return msg




#Do not modify the following:
message = "Hide this SECRET message."
message2 = "THIS is another secret MESSAGE."

print(encrypt(message)=='H*d* th*s S*CR*T m*ss*g*.')
print(encrypt(message2)=='TH*S *s *n*th*r s*cr*t M*SS*G*.')
```

```python
#Q8 Alternative solution

def encrypt(secret):
    for char in secret:
        if char.lower() in 'aeiou':
            secret = secret.replace(char, "*")
    return secret


#Do not modify the following:
message = "Hide this SECRET message."
message2 = "THIS is another secret MESSAGE."

print(encrypt(message)=='H*d* th*s S*CR*T m*ss*g*.')
print(encrypt(message2)=='TH*S *s *n*th*r s*cr*t M*SS*G*.')
```
