```python
#Task 3.2

def insertion_sort(lst):   
    for i in range(len(lst)-1):
        if get_temp(lst[i])>get_temp(lst[i+1]):
            lst[i] , lst[i+1] = lst[i+1], lst[i]
            insertion_sort(lst)
        else:
            pass
    return lst
```

```python
# Task 5.1

def hash(email):
    at = email.index('@')
    name = email[:at]
    sum = 0 
    for i in range(len(name)):
        sum += ord(name[i])
    return sum %20
        




hash('abc@yijc.edu.sg')
```

```python
# Task 5.2

def hashtable(lst):
    hashtable = [''] * 20
    for email in lst:
        currentemailhash = hash(email)
        if hashtable[currentemailhash] != '':
            print('its probing time')
            not_found = True
            probs = 0
            while not_found and probs < 20:
                probs +=1
                if hashtable[(currentemailhash+(probs)*2)%20] == '':
                    not_found = False
                    hashtable[(currentemailhash+(probs)*2)%20] = email
                    break
                
        else:
            hashtable[currentemailhash] = email
    return hashtable        
```
