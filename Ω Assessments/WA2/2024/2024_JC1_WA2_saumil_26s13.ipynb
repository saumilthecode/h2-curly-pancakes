# Question 5: Stack ADT - Check for Valid Tower [5 marks]

```python
def checktower(seq):
    stack = []
    for char in seq:
        if char == "P":
            if not stack:
                return False
            stack.pop()
        elif char.isdigit():
            disk = int(char)
            if stack and disk >= stack[-1]:
                return False
            stack.append(disk)
        else:
            return False

    return True
```

# Question 6: Stack ADT - Back Tracking [5 marks]

```python
def backtrack(seq):
    stack = []
    for item in seq:
        stack.append(item)
    seq = ""
    while stack:
        letter = stack.pop()
        if letter == 'N':
            seq = seq + 'S'
        elif letter == 'E':
            seq = seq + 'W'
        elif letter == 'S':
            seq = seq + 'N'
        elif letter == 'W':
            seq = seq + 'E'
    return seq
        
    

backtrack('SSSE')



```
