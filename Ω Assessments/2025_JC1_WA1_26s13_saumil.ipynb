# Q2 Postal Code Check (5 marks)

```python
#Q2

# If it starts with 01 to 15, output "This is the Central Region."
# If it starts with 16 to 41, output "This is the East Region."
# If it starts with 42 to 52 or 80 to 99, output "This is the North-East Region."
# If it starts with 53 to 57 or 72 to 78, output "This is the North Region."
# If it starts with 58 to 71, output "This is the West Region."
# Otherwise, output "Region not classified."


postal = input('enter a 6-digit Singapore postal code')
length = len(postal)


if length != 6 and not postal.isdigit():
    print('Invalid input')
else:
    first2 = postal[0:2]
    first2 = int(first2)
    if first2 <= 15 and first2 >= 1:
        print('This is the Central Region.')
    elif first2 <= 41 and first2>= 16:
        print('This is the East Region.')
    elif (first2 <= 52 and first2>= 42) or (first2<=99 and first2>=80):
        print("This is the North-East Region.")
    elif (first2 <= 57 and first2>= 53) or (first2<=78 and first2>=72):
        print("This is the North Region.")
    elif first2 <= 71 and first2>= 58:
        print("This is the West Region.")
    else:
        print("Region not classified.")

    


```

# Q4 Area of Triangle (2 marks)

```python
#Q4

def area_triangle(a, b, c):
    s = (a+b+c)/2
    area = (s*(s-a)*(s-b)*(s-c))**0.5
    return round(area,2)
    
area_triangle(3, 4, 5) 


```

# Q5 Classify the Temperature (3 marks)

```python
#Q5

def classify(temp):
    if temp <0:
        return 'Freezing'
    elif 0 <= temp <10:
        return 'Cold'
    elif 10 <= temp < 20:
        return 'Cool'
    elif 20 <= temp < 30:
        return 'Warm'
    else:
        return 'Hot'

```

# Q7 Modify a Tuple (5 marks)

```python
#Q7

def modify(data, new):
    final_tupl = ()
    for tupl in data:
        if tupl[0] == new[0]:
            final_tupl += (new,)
        else:
            final_tupl += (tupl,)
    return final_tupl
        
            






#Do not modify the following:
students = (("Alice", 160), ("Bob", 175), ("Charlie", 168), ("Diana", 170), ("Ethan", 165))
charlie = ("Charlie", 170)

modify(students,charlie)
```

# Q8 Encrypt a Secret Message (5 marks)

```python
#Q8

def encrypt(secret):
    newstr = ''
    down = secret.lower()
    banned = ('a','e','i','o','u')
    for i in range(len(down)):
        if down[i] in banned:
            newstr += '*'
        else:
            newstr += secret[i]
            print(newstr)
    return newstr
        
        






#Do not modify the following:
message = "Hide this SECRET message."
encrypt(message)


```
