# Question 1 - Recursion [4 marks]

```python
def remove_digits(text):
    if text == '':                                      #[1m] handle the base case
        return ''
    if text[0] in '0123456789':                         #[1m] recognise and remove a digit
        return remove_digits(text[1:])                  #[1m] recurse using a smaller string
    else:
        return text[0] + remove_digits(text[1:])         #[1m] retain other characters in order
```

# Question 3 - Stack ADT [4 marks]

```python
def create_stk():          # 1m create an empty stack
    return []

def push(stk, ele):        
    stk.append(ele)        # 1m use append() for push()   
                           
def pop(stk):
    if stk != []: 
        return stk.pop()   # 1m remove and return popped value
    else:                  # 0m if not checking empty stack
        return None

def is_empty(stk):         # 1m return True/False
    return len(stk) == 0
```

# Question 4 - Application of Stack [6 marks]

```python
def separate(string):
    stk = create_stk()
    output = ''
    for ele in string:                  #1m iterate string
        if ele.isdigit():               #1m check digit/char
            output = output + ele       #1m concatenate digit
        else:
            push(stk,ele)               #1m use stack to collect char

    temp = ''
    while not is_empty(stk):
        temp = pop(stk) + temp          #1m pop char out

    return output + temp                #1m correct order of digit/char



print(separate('62J9M1AG61'))
```
