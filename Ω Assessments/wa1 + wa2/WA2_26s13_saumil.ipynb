## Task 1.1 Remove the Latest Character [3 marks]


```python
def remove_last(text):
    if len(text) <1:
        return text
    if len(text) == 1:
        return ''
    return text[0] + remove_last(text[1:])
```

## Task 1.2 Multiple Backspacing [4 marks]


```python
# Supplied so that you can attempt Task 1.2 even if you were unable to complete Task 1.1.
def remove_last(text):
    return text[:-1]

## Do not edit above this line
###########################

def edit_text(text):
    if text == '':
        return text
    if len(text) == 1:
        if text == '<':
            return ''
        return text
    if text[-1] == '<':
        return remove_last(edit_text(text[:-1]))
    else:
        return edit_text(text[:-1]) + text[-1]


edit_text("ROBOT")
# edit_text("CART<")
edit_text("Room 2B!<<")
```

## Task 2.2 Application of Stack ADT [4 marks]


```python
# Supplied Stack ADT for this task
def make_empty_stack():
    return []


def push(stack, item):
    stack.append(item)


def pop(stack):
    if len(stack) == 0:
        return None
    return stack.pop()


def is_empty(stack):
    return len(stack) == 0


## Do not edit above this line
###########################

def edit_text(text):
    stk = make_empty_stack()
    for char in text:
        push(stk,char)
        if char == '<':
            pop(stk)
            pop(stk)
    return ''.join(stk)

edit_text("ROBOT")
```

## Task 3.2 Boarding a Ride [5 marks]


```python
from queue_adt import *

# ride_q2 is provided so that students who cannot do Task 3.1 can continue.
ride_q2 = [
    ("Group1", 3),
    ("Group2", 2),
    ("Group3", 4),
    ("Group4", 3),
]


## Do not edit above this line
###########################

def board_ride(queue, n):
    boarded_groups = []
    ppl = n
    for i in range(len(queue)):
        firstgrp = front(queue)
        if firstgrp[1] <= ppl:
            ppl -= firstgrp[1]
            boarded_groups.append(front(queue)[0])
            dequeue(queue)
        else:
            break
    return [boarded_groups,ppl]
board_ride(make_queue((("Group1", 3), ("Group2", 2), ("Group3", 4), ("Group4", 3))), 11)
```

```python
# Public tests
test_q = make_queue(ride_q2)
returned = board_ride(test_q, 6)
print("board_ride return value")
print("Expected:", [["Group1", "Group2"], 1])
print("Actual:  ", returned)

print()
print("Groups remaining in the test queue")
print("Expected:", [("Group3", 4), ("Group4", 3)])
print("Actual:  ", test_q)

# Add more tests of your own below — vary the seats and group sizes!
```

## Task 4.1 Read Data from a CSV Text File [6 marks]


```python
def readfile(filename):
    filler = ()
    with open(filename) as f:
        lines = f.readlines()
        f.close()
    for line in lines[1:]:
        line = line.strip()
        numbr, recipient, (weight) = line.split(',')
        filler += ((numbr, recipient, float(weight)),)
        
    return filler
readfile("DATA.csv")
```

## Task 4.2 Hash Function [3 marks]


```python
def hash(string, n):
    hashervalue = 0
    for char in string:
        hashervalue += ord(char)
    return hashervalue%n
# Initial test case
# hash("P104", 13)  # Expected: 8
```
