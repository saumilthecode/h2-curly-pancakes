### Question 2: Calculate the Area of a Parallelogram [2 marks]

```python
from math import *

def area_calc(side1, side2, angle_in_degrees):
    angle_in_radians = (angle_in_degrees / 180) * pi
    return side1*side2*sin(angle_in_radians)



```

### Question 4: Insert at symbol [5 marks]

```python
def insert(text, sentence):
    replaced = ''
    for i in range(len(sentence)):
        if sentence[i] == '%':
            replaced += text
        else:
            replaced += sentence[i]
    return replaced

insert("John Doe", "The medical records of % have been submitted.")
```

```python
#Test Case for Q4
print(insert("John Doe", "The medical records of % have been submitted.") == "The medical records of John Doe have been submitted.")
```

### Question 6: Sum the Lucas Numbers [5 marks]

```python
def lucas(x):
    if x <= 0:
        return 0
    elif x == 1:
        return 2
    elif x == 2:
        return 1
    return lucas(x - 1) + lucas(x - 2)
    
# Do not modify the Lucas(x) function.


#write your program code below:
def sum_lucas(n):
    bool = False
    sum  = 0
    if n > 2:
        while bool != True:
            for i in range(100000):
                if lucas(i) > n:
                    bool = True
                    return sum
                else:
                    sum += lucas(i)
sum_lucas(4)
```

```python
#Test cases for Q6
print(sum_lucas(4) == 10)
print(sum_lucas(8) == 17)
```

### Question 7: Searching in a tuple [4 marks]

```python
def search(data, name):
    sum = 0
    for ele in data:
        if ele[0] == name:
            return ele
        else:
            sum += 1
            print(sum)
            if sum >= len(data):
                return False
    


# Do not modify the following:
customers = (("Jane", 21, "F"),("Jian Hong", 19, "M"),("Peter Toh", 23, "M"),("Adam", 18, "M"))
# len(customers)
search(customers, 'Jane')
```

### Question 8: Body Mass Index Calculator [3 marks]

```python
def bmi(weight, height):
    height = height/100
    mi = weight/(height**2)
    return round(mi,1)
bmi(89, 180)
```

### Question 9: BMI Classifier [7 marks]

```python
def status(data):        
    response = ()
    status = ''
    for student in data:
        mi = bmi(student[1], student[2])
        if mi < 18.5:
            status = 'Underweight'
        elif mi>=18.5 and mi <= 25.0:
            status = 'Normal Weight'
        elif mi > 25.0:
            status = 'Overweight'
        response = response + ((student[0],mi,status),)

    return response



# Do not modify the following:
students = (('Andy', 51.3, 166), ('Bobby', 75.9, 172), ('Charlie', 69.4, 176), 
            ('Daniel', 64.6, 173), ('Edgar', 74.5, 172), ('Felix', 55.9, 174), 
            ('George', 64.2, 177), ('Hubert', 61.9, 178), ('Ivan', 53, 150), ('John', 51.7, 170))
        
```

```python
#Test Case for Q9
output = status(students)
print(output[0] == ('Andy', 18.6, 'Normal Weight'))
```
