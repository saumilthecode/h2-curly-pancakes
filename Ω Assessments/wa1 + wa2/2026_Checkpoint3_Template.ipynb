## Question 3 : Increasing the Values in a Simple Dictionary

Write a function `increase_by_one` which takes in a dictionary and modifies the dictionary by increasing all values in it by 1. 


```python
def increase_by_one(d):
    for key in d:
        d[key] +=1
    return d
        



increase_by_one({'first':27, 'second':16, 'third':8})

```

## Question 4 : Combine 2 Dictionaries

Write a function `combine(d1,d2)` to return a new dictionary combining all the items in the dictionaries.


```python
def combine(d1, d2):
    final_dict = {}
    for key in d1:
        if key in final_dict:          
            final_dict[key] += d1[key]
        else:
            final_dict[key] = d1[key]  
    for key in d2:
        if key in final_dict:          
            final_dict[key] += d2[key]
        else:
            final_dict[key] = d2[key] 
            
    return final_dict
        
# Do not modify the following:
John = {'burger':2, 'ice milo':1, 'fries':1}
Mary = {'burger':1, 'coke':1, 'apple pie':2, 'fries':3}

combine(John, Mary)
```

## Question 5 : Upsize Player's Inventory by Looting

Write a function `add_inv(inv, items)`, where the `inv` parameter is a dictionary representing the player's inventory and `items` is a list containing items to be added to the inventory.

The function should return the player's upsized inventory `inv` with the list `items` emptied.   

```python
def add_inv(inv, items):
    for item in items:
        if item in inv:
            inv[item] +=1
        else:
            inv[item] = 1
    items.clear()       
    return inv

# Do not modify the following:
my_bag = {'coin': 2, 'health_pot': 3, 'mana_pot': 5, 'scroll': 2, 'dagger': 1, 'ore': 3, 'mace': 1, 'gloves': 1}
barlog_loot = [ 'leather', 'sword', 'scroll', 'scroll', 'coin', 'coin', 'helmet' ]


add_inv(my_bag, barlog_loot)
```
