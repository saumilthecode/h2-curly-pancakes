#### Question 7: An Application of Stack ADT [5 marks]

```python
from LA2_module import *
#### Do not edit above this line #####

def merge(seq1, seq2):
    stk1 = make_stack(seq1)
    stk2 = make_stack(seq2)
    
    result = make_empty_stack()

    while stk1 and stk2:  
        if peek(stk1) <= peek(stk2):
            push(result,pop(stk1))
        else:
            push(result,pop(stk2))

    if is_empty_stack(stk1):
        while stk2:
            push(result,pop(stk2))

    else:
        while stk1:
            push(result,pop(stk1))

    return result
merge('YUSNJIHC','VONNIA')
```
