```python
def div19(num):
    not_ended = True
    while not_ended:
        if num > 19:
            print('Number:', num)
            string_numb = str(num)
            length = len(string_numb)
            first_few = string_numb[:length-1]
            last = string_numb[length-1]
            print('Firstfew:',first_few , ' Last:' , last)
            num = int(first_few) + (2 * int(last))
            
        elif num ==19:
            return True
        else:
            return False
        


div19(82099)
```
