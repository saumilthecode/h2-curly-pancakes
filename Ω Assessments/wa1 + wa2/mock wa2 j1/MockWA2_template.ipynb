### Solution 1 - Library loans summary

````python
def summarise_late_loans(filename):
    total_late_days = 0
    late_students = []
    late_days_by_book = {}

    f = open(filename, "r")
    header = f.readline()
    line = f.readline()

    while line != "":
        line = line.strip()
        parts = line.split(",")

        student = parts[0]
        book_code = parts[2]
        days_late = int(parts[3])

        total_late_days = total_late_days + days_late

        if days_late > 0:
            late_students.append(student)

        if book_code not in late_days_by_book:
            late_days_by_book[book_code] = days_late
        else:
            late_days_by_book[book_code] = late_days_by_book[book_code] + days_late

        line = f.readline()

    f.close()
    return (total_late_days, late_students, late_days_by_book)


summary = summarise_late_loans("library_loans.csv")
print(isinstance(summary, tuple) and len(summary) == 3 and summary[0] == 23)
print(isinstance(summary, tuple) and len(summary) == 3 and summary[1] == ["Bella", "Chen Wei", "Ethan", "Farah", "Harun", "Irfan", "Jia En"])
print(isinstance(summary, tuple) and len(summary) == 3 and summary[2].get("BK205") == 10)
````

### Solution 2 - Student ID hash table

````python
def build_student_hash_table(filename, table_size):
    hash_table = [None] * table_size

    f = open(filename, "r")
    header = f.readline()
    line = f.readline()

    while line != "":
        line = line.strip()
        parts = line.split(",")

        student_id = int(parts[0])
        name = parts[1]
        index = student_id % table_size

        if hash_table[index] is None:
            hash_table[index] = (student_id, name)
        else:
            hash_table[index] = "COLLISION"

        line = f.readline()

    f.close()
    return hash_table


student_table = build_student_hash_table("student_ids.csv", 10)
print(isinstance(student_table, list) and len(student_table) == 10)
print(isinstance(student_table, list) and student_table[2] == (102, "Amir"))
print(isinstance(student_table, list) and student_table[4] == (894, "Harun"))
print(isinstance(student_table, list) and student_table[7] == "COLLISION")
````

### Solution 3 - Stack application

````python
def remove_adjacent_pairs(text):
    stack = []

    for character in text:
        if len(stack) > 0 and stack[-1] == character:
            stack.pop()
        else:
            stack.append(character)

    result = ""

    for character in stack:
        result = result + character

    return result


print(remove_adjacent_pairs("abbaca") == "ca")
print(remove_adjacent_pairs("aabb") == "")
print(remove_adjacent_pairs("") == "")
````

### Solution 4 - Queue application

````python
from queue_adt import make_empty_queue, enqueue, dequeue, is_empty


def serve_requests(requests, number_to_serve):
    queue = make_empty_queue()

    for request in requests:
        enqueue(queue, request)

    served = []
    count = 0

    while count < number_to_serve and not is_empty(queue):
        served.append(dequeue(queue))
        count = count + 1

    remaining = []

    while not is_empty(queue):
        remaining.append(dequeue(queue))

    return (served, remaining)


print(serve_requests(["A", "B", "C"], 2) == (["A", "B"], ["C"]))
print(serve_requests(["A"], 3) == (["A"], []))
print(serve_requests([], 2) == ([], []))
````

### Solution 5 - Recursive digit sum

````python
def digit_sum(n):
    if n < 10:
        return n
    else:
        last_digit = n % 10
        remaining_digits = n // 10
        return last_digit + digit_sum(remaining_digits)


print(digit_sum(507) == 12)
print(digit_sum(0) == 0)
print(digit_sum(9999) == 36)
print(digit_sum(12345) == 15)
````
