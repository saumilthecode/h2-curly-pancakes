# Task 2.2 Compute Delivery Charge [5 marks]

```python
def delivery_charge(length, width, height):
    volume = length*width*height
    weight = volume/6000
    if 0< volume <= 140000:
        cost = (weight*15)
        return(round(cost))
    elif 14000<volume <= 30000:
        cost = (weight*10)
        return(round(cost))
    elif 30000 < volume <= 70000:
        cost = (weight*8)
        return(round(cost))
    else:
        print('Parcel is too big for delivery')
        return None

               



print(delivery_charge(35,25,10))    
print(delivery_charge(40,30,15))    
print(delivery_charge(50,40,25))    
```

# Task 4 Storing Data in a Tuple [4 marks]

```python
def compute(results):
    tuplreturn = ()
    sum = 0
    numb = 0
    for s in results:
        for i in range(len(s)):
            if type(s[i])== int :
                sum += s[i]
                numb += 1
        tuplreturn += ((s[0],round((sum/numb),2)),)
        sum = 0 
        numb = 0
    return tuplreturn
                


class123 = (('Amy', 57, 69, 52), ('Bob', 44, 45, 53), ('Charlie', 84, 72, 68))

compute(class123)
```
