```python
def bowling_score(params):
    score = 0
    frame = 0
    while params != "" and frame < 10:
        if params[0] == "X":
            score += 10
            if params[1] == "X":
                score += 10
            else:
                score += int(params[1])
            if params[2] == "X":
                score += 10
            else:
                score += int(params[2])
            params = params[1:]
            frame += 1
        else:
            first = int(params[0])
            if params[1] == "X":
                second = 10
            else:
                second = int(params[1])
            if first + second == 10:
                score += 10
                if params[2] == "X":
                    score += 10
                else:
                    score += int(params[2])
            else:
                score += first + second
            params = params[2:]
            frame += 1
    return score

print(bowling_score("00000000000000000000"))  # 0
print(bowling_score("21212121212121212121"))  # 30
print(bowling_score("XX1200000000000000"))  # 37
print(bowling_score("0000XX120000000000"))  # 37
print(bowling_score("0000XXX1234000000"))  # 74
print(bowling_score("00007382120000000000"))  # 32
print(bowling_score("000000000000000000731"))  # 11
print(bowling_score("000000000000000000X12"))  # 13
print(bowling_score("XXXXXXXXXXXX"))  # 300
print(bowling_score("000000000000000000XXX"))  # 30
print(bowling_score("X7390X088206XXX81"))  # 167
print(bowling_score("X37X28X64X55X46X"))  # 200
print(bowling_score("37X28X64X55X46X28"))  # 200
print(bowling_score("0X0X0X0X0X0X0X0X0X0XX"))  # 110
```
