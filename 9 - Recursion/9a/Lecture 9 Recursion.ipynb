```python
def factorial(n):
    if n == 1: #Base Case
        return 1
    else:      #General Case
        return n * factorial(n - 1)
```

```python
def ping(n):
    if n == 0:
        return n
    else:
        print("Ping!")
        pong(n - 1)

def pong(n):
    if n == 0:
        return n
    else:
        print("Pong!")
        ping(n - 1)
```

## Q10

```python
def count_odd(n):
    if n == 1:
        return 1
    if n % 2 == 1:
        print(str(n) + 'is odd')
        return n + count_odd(n-1)
    else:
        print(str(n) + 'is even')
        return count_odd(n-1)

count_odd(2)
```

## Q11

```python
def count_even(n):
    if n == 1:
        return 0
    if n % 2 == 1:
        print(str(n) + 'is odd')
        return count_even(n-1)
    else:
        print(str(n) + 'is even')
        return n + count_even(n-1)

count_even(1)
```
