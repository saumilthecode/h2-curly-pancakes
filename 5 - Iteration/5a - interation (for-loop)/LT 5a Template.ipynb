### Question 4: factorial(n)

```python
def factorial(n):
    factorial = 1
    for i in range (1,n+1):
        factorial = factorial * i
    return factorial
```

```python
factorial(4)  #This should give 1 * 2 * 3 * 4 = 24
```

### Question 5: compute(x, n)

```python
def compute(x, n): # x ^n
    if n ==0:
        return 1
    else:
        result = x
        for i in range(1,n):
            result *= x
        return result
```

```python
print(compute(6,4))  #This should return 1296
print(compute(2,3)) #8
```

### Question 7: sum_odd(n)

```python
def sum_odd(n):
    sum = 0
    for i in range (n+1):
        if (i % 2) == 1:
            sum = sum + i
    return sum
    
sum_odd(5) #9
```

```python
sum_odd(5)   #This should return 1 + 3 + 5 which is 9
```

### Question 8: Sum of first n squares (Iterative)

```python
def sum_n_squares(n):
    sum = 0
    for i in range(1,n+1):
        square = i **2
        sum += square
    return sum
```

```python
sum_n_squares(5)  #This should return 55
```

### Question 9: Counting a character in the string

```python
def count_char(string, character):
    counter = 0
    for i in string:
        if i == character:
            counter = counter +1
        else:
            pass
    return counter
        
```

```python
count_char('banana','a')  #There are 3 a's in banana
```

### Question 11: Replace all the occurrences of a letter in the string

```python
def replace_letter(msg, letter, new_char):
    string = ''
    for i in msg:
        if i == letter:
            string += new_char
        else:
            string += i

    return string
```

```python
print(replace_letter("31242154125", "1", "0"))     # "30242054025"
print(replace_letter("thisisasong", "s", "e"))     # "thieieaeong"
```
