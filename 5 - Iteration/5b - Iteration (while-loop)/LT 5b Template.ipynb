```python
def print_to(n):
    number = 0
    while number <= n:
        print(number)
        number +=1
```

### Question 5: compute(x, n)

```python
def compute(x, n):
    total = 1
    while n > 0:
        total *= x
        n -= 1
    return total
```

```python
compute(2,3)   #This should return 8
```

### Question 6: powers_of_two(n)

```python
def powers_of_two(n):
    power = 0
    while n != 1:
        n = n/2
        power +=1
    
    return power
```

```python
powers_of_two(16) #this should give 4 since you can divide 16 by 2 4 times.
```
