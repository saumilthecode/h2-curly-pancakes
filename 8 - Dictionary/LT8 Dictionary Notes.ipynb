- The built-in function dict(…) applies to any sequence of key-value pairs :

```python
t1 = ((1, 2), (2, 4), (3, 6))   #tuple of tuples
t2 = [(1, 2), (2, 4), (3, 6)]   #list of tuples
t3 = ([1, 2], [2, 4], [3, 6])   #tuple of lists
t4 = [[1, 2], [2, 4], [3, 6]]   #list of lists

print(t1)
print(t2)
print(t3)
print(t4)
print('\n')
print(dict(t1))
print(dict(t2))
print(dict(t3))
print(dict(t4))
```

### Output the keys and values into lists 

```python
s =(('soup', (('mushroom', 4),('tomato',3))),
    ('mains', ('chicken',10)), 
    ('dessert',{'weekdays':0 , 'weekends':('lava cake',7)}))

menu = dict(s) 
```

```python
list(menu.keys())
```

```python
for i in range(len(menu.keys())):
    print(i)
```

```python
list(menu.values())
```

### Iterate through a Dictionary

```python
for key in menu:                 # by default, looping a dictionary will iterate by key
    print(key)
```

```python
for key in menu:
    print(menu[key])             # and you can useg the key to access the value
```

```python
for key, value in menu.items():  # using .items() will allow you to access both at the same time
    print(key)
```

```python
for key, value in menu.items():
    print(value)
```

# Question 14

```python
def count(seq):
    uniques = []
    tupl = {}
    for i in seq:
        if i not in uniques:
            uniques += [i,]
            tupl[i] = 0
        if i in uniques:
            tupl[i] +=1
    return tupl



lst = [1, 2, 3, 1, 1, 2, 3, 1, 1, 3, 2]
lst2 = ['a', 'b', 'c', 'a', 'a', 'b', 'c', 'a', 'a', 'c', 'b']
count(lst)
```

# Question 15

```python
def count_words(paragraph):
    words = paragraph.split()
    counts = {}

    for word in words:
        if word not in counts:
            counts[word] = 0
        counts[word] += 1

    return counts
    
twister = '''Peter Piper picked a peck of pickled peppers
a peck of pickled peppers Peter Piper picked
if Peter Piper picked a peck of pickled peppers
where’s the peck of pickled peppers Peter Piper picked'''

count_words(twister)
```

# Question 16

```python
def count_vowels(word):
    vowels = "aeiou"
    counts = {}

    for char in word.lower():
        if char in vowels:
            if char not in counts:
                counts[char] = 0
            counts[char] += 1

    return counts

count_vowels('apple')
```

```python
def average(result1, result2):
    d = {}
    for subject in result1:
        average = (result1[subject] + result2[subject])/2
        d[subject] = average
        
    return d





# Do not modify the following:
studentA_CT1 = {'H1GP': 58, 'H2MA': 66, 'H2CP': 75, 'H2PH': 55, 'H1EC': 45}
studentA_CT2 = {'H1GP': 62, 'H2MA': 73, 'H2CP': 72, 'H2PH': 61, 'H1EC': 47}

studentB_CT1 = {'H1GP': 60, 'H2MA': 76, 'H2EC': 53, 'H2PH': 66, 'H2CH': 59}
studentB_CT2 = {'H1GP': 62, 'H2MA': 73, 'H2EC': 45, 'H2PH': 61, 'H2CH': 57}

average(studentA_CT1,studentA_CT2)
```
