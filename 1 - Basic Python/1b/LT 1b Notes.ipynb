A string is any sequence of characters. We usually use them to store words, sentences or even entire paragraphs. The data type of a string is __str__.

# String Slicing

Each character in a string corresponds to an __index number, starting with 0__. We can find a character that corresponds to a particular index within a string by using the index operator `[]`.

```python
s = 'abcd'

print(s[0])   # returns the character at index 0
print(s[2])   # returns the character at index 2
```

We can also return a range of characters by using the slice syntax `[start:stop]`. For examples:

```python
string = 'HelloWorld'

print(string[2:5])   # returns the characters from index 2 (inclusive) to index 5 (exclusive)
print(string[2:])    # returns the characters from index 2 (inclusive) onwards, if the stop index is not specified
print(string[:5])    # returns the characters before index 5 (exclusive), if the start index is not specified
```

We may even add the third argument called `step` in our slice syntax such that it becomes `[start:stop:step]`. This is an optional argument that determines the increment between each index for slicing. When not specified, its default value is 1.

```python
string = 'HelloWorld'

print(string[2:8])     # starting from index 2 until index 7, with increment of 1 (default) for each step
print(string[2:8:2])   # starting from index 2 until index 7, with increment of 2 for each step
print(string[2:8:3])   # starting from index 2 until index 7, with increment of 3 for each step
```

It is important to note that the `stop` index is exclusive. You can remember this by noting that you should put a cursor each before the `start` and `stop` index positions respectively. Then the sliced section will be between the pair of cursors.

# Negative Indexing

Python programming language also supports negative indexing, something which is not available in most other programming languages. This means that the index value of -1 gives the last character, -2 gives the second last character of a string and so on.

```python
new_string = 'Singapore'

print(new_string[-1])
print(new_string[-3])
```

Although it is possible to use negative indexing in string slicing, it may be quite difficult to understand how it works and you are discouraged from using it. You may still explore it on your own if you are interested.

However, using a __negative step__ can be quite useful sometimes as it would allow us to __reverse__ a string. 

```python
new_string = 'Singapore'

print(new_string[::-1])   # start & stop arguments are not specified to get the whole string
print(new_string[::-2])   # starting from the last index to the first index, with an increment of 2
```
