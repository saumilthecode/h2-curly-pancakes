# Data Type

Python support many basic data types. The common ones are
- __Integer (int):__ ..., -100, -2, 0, 1, 2, 300, ...
- __Floating-point number (float):__ ..., -3.1415, -1.0, 0.00001, 23.0, ...
- __String (str):__ 'Hi', "Hello", '', ...
- __Boolean (bool):__ True, False
- __None (NoneType)__

To check the type of a data, we may use the `type()` function. Refer to the examples below.

```python
print(type(123))
print(type(12.3))
print(type('123'))
print(type(True))
print(type(None))
```

We can usually convert one data type into another, as illustrated in the examples below.

```python
x = str(123)     # 123 is an integer
print(type(x))

y = float('45.2')     # '45.2' is a string
print(type(y))

z = int('yijc')     # Can a string be converted into an integer? 
```

Notice that we have used comments (by using the hash character `#`) in the examples above. Comments are often useful in making our codes more readable.

Some type conversions are not allowed, as shown in the last example. You should explore them on your own.

# Arithmetic Operators

Arithmetic operators are used to perform mathematical operations like addition, substraction, multiplication, division, etc.

| Operator | Meaning | Example |
| - | - | - |
| + | Add two operator | x + y |
| - | Substract right operand from the left | x - y |
| * | Multiply two operands | x * y |
| / | Divide left operand by the right one (result in float) | x / y |
| % | Modulus - remainder of the division of left operand by the right | x % y |
| // | Floor division discards fractional part | x // y |
| ** | Exponent - left operand raised to the power of the right | x ** y |

Here are some examples:

```python
print(2+10)
print(2-10)
print(2*10)
print(10/2)
print(10%3)
print(10//3)
print(2**3)
```
