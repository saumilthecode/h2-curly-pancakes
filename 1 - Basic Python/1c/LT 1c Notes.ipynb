# Boolean

The Boolean data type is named after the mathematician __George Boole__. It has only two values: `True` and `False`.

In Python version 3 and above, `True` and `False` will always be equal to __1__ and __0__ respectively. Moreover, anything that is __not 0__ or __not empty__ will also be evaluated as `True`.

```python
print(bool(2))
print(bool(-5.3))
print(bool(0.0))
print(bool('Hello'))
print(bool(''))

print("\n")


# print("big" > "small")


False >= 8
```

# Comparison Operators

- Relational operators are used to compare Boolean values.
- It either returns True of False according to the satified condition.

| Operator | Meaning | Example |
| - | - | - |
| > | Greater than - `True` if left operand is greater than the right | x > y |
| < | Less than - `True` if left operand is less than the right | x < y |
| == | Equal to - `True` if both operands have the same value | x == y |
| != | Not equal to - `True` if both operands do not have the same value | x != y |
| >= | Greater than or equal to - `True` if left operand is greater than or equal to the right | x >= y |
| <= | Less than or equal to - `True` if left operand is less than or equal to the right | x <= y |

```python
print(1 > 2)
print(1 < 2)
print(5 > 5.0)
print(5 <= 5.0)
print(5 == 5.0)
print(5 != 5.0)
```

__Notes:__
1. From the last 3 examples, observe that the integer 5 and the float 5.0 are treated as equal.
2. There is a difference between the comparison operator `==` and the assignment operator `=`.

# Boolean Operators

- Boolean operators are used to compare logical operands.
- It either returns `True` or `False` according to the satisfied condition.

| Operator | Meaning | Example |
| - | - | - |
| and | `True` if both of the operands are `True` | x and y |
| or | `True` if either of the operands is `True` | x or y |
| not | `True` if the operand is `False` (negate the operand) | not x |

```python
x = 10
print(x>1 and x<20)
print(x>11 or x<20)
print(not x>0)
```
