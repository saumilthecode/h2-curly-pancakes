# Q1 Linear Search Recursive

```python
def find(seq, item):
    if len(seq) == 0:
        return False
    elif seq[0] == item:
        return True
    else:
        return find(seq[1:], item)
        
def index(seq, item):
    if len(seq) == 0:
        print("Item not found.")
        return 0
    elif seq[0] == item:
        return 0
    else:
        return 1 + index(seq[1:], item)

def count(seq, item):
    if len(seq) == 0:
        return 0
    elif seq[0] == item:
        return 1 + count(seq[1:], item)
    else:
        return count(seq[1:], item)

# Do not remove:
string = 'mississippi'
```
