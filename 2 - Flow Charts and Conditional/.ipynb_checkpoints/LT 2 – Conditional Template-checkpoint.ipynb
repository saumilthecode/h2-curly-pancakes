## Question 4

```python
name = input('Enter a name : ')

if name == 'Alice':
    print('Hi, Alice.')
else:
    age = input('Enter the age : ')
    if age < 12 :
        print ('You are not Alice, kiddo.')
    else:
        print('You are neither Alice nor a little kid.')
```

## Question 5

```python
name = input('Enter a name : ')

if name == 'Alice':
    print('Hi, Alice.')
else:
    #name is not alice
    age = int(input('Enter the age : ')) #asume that user is a nice person who entered a number
    if age < 12 :
        print('You are not Alice, Kiddo.')
    else: #assume that user has entered a number
        if age > 2000:
            print('Unlike you, Alice is not an undead, immortal vampire')
        else:
            if age > 100:
                print('You are not Alice grannie.')
            else:
                exit()
```

## Question 6

```python
n1 = input('Enter a number for n1: ')  #input is always string
n2 = input('Enter a number for n2: ')
char = input('Enter "+", "-", or "*": ')

n1 = float(n1)
n2 = float(n2)

if char == '+':
    print(n1 + n2)
elif char == "-":
    print( n2 - n1)
elif char == "*":
    print(n1 * n2)
else:
    print("invalid character")
```

## Question 7

```python
#Take inputs for n1 and n2
n1 = float(input('Enter a number for n1: '))
n2 = float(input('Enter a number for n2: '))

#check if n2 == 0
if n2 == 0:
    print("Division by 0 is not allowed")
    n2 = float(input("Please re-enter n2:"))
    if n2 == 0:
        print('ok no more for you')
    
else:
    #calc quotient
    quotient = (n1//n2)
    #calc remainder
    remainder = (n1/n2) - quotient
    #print quotient, remainder
    print('your quotient is ' + str(quotient) + "\nyour remainder is " + str(remainder))
```
