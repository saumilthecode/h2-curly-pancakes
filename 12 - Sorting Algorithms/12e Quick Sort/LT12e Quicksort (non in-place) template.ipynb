# Quick Sort (non in-place) Algorithm

## Description :

1. Use the last element as the `pivot`.
2. Iterate through the sequence of elements and append all the smaller values into the `left` list and the larger values into the `right` list. 
3. Join the `left`, `pivot` and `right` into a single list.
4. Recursively do step 1-2 for the `left` and the `right` lists until there is only one or no element in the list.
5. This process using extra memory to create some lists to perform the quicksort, hence it is a non in-place sorting algorithm.
