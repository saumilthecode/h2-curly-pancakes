## Q7 Core Skill 2: Swap adjacent elements down a sequence in a single pass

```python
def single_pass(seq):
    for i in range(len(seq) - 1):
        if seq[i] > seq[i + 1]:
            seq[i], seq[i + 1] = seq[i + 1], seq[i]
```

## Q8 Simple Bubble Sort Algorithm

```python
def simple_bubblesort(seq):
    for ijk in range(len(seq)):
        for i in range(len(seq) - 1):
            if seq[i] > seq[i + 1]:
                seq[i], seq[i + 1] = seq[i + 1], seq[i]
    return seq


lst = [5, 4, 3, 2, 1, 0]

print(simple_bubblesort(lst))
```
