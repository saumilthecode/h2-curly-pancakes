## Q1 Pseudocode: Insertion Sort

```/code
1 FOR Pointer <- 2 TO NumberOfitems
2     ItemToBeInserted <- MyList[Pointer]
3     CurrentItem <- Pointer -1
4     WHILE (MyList[CurrentItem] > ItemToBeInserted) AND (CurrentITEM > 0)
5         MyList[CurrentItem + 1]  <- MyList[CurrentItem]
6         CurrentItem <- CurrentItem -1
7     ENDWHILE
8     MyList[CurrentItem + 1] <- ItemToBeInserted
9 ENDFOR
```

### Note 1:
- `MyList[1]`, where Pointer=1, is the first element.
- Python uses zero-based indexing and `lst[0]` is the first element.

- For Line 1: `Pointer <- 2 TO NumberOfitems` becomes `range(1, len(lst))`.

### Note 2:
- `CurrentItem` is the position of an element in `MyList`, the smallest value is 1 which is the first element.
- To iterate all the way to the first element in the sort sub-list, the index of `CurrentItem` should go all the way down to 0.
- Hence for Line 4: `CurrentITEM > 0` becomes `CurrentITEM >= 0`.

````python
#Using the Pseudocode:
def insertion_sort(lst):
    for pointer in range(1, len(lst)):
        item_to_be_inserted = lst[pointer]
        current_item = pointer - 1

        while current_item >= 0 and lst[current_item] > item_to_be_inserted:
            lst[current_item + 1] = lst[current_item]
            current_item -= 1

        lst[current_item + 1] = item_to_be_inserted

    return lst

insertion_sort([7,1,3,5,8,4])
````

## Q4 Insertion Sort

````python
def insertion_sort(lst):
    for k in range(1,len(lst)):
        i = k         #item = lst[i]
        while lst[i-1] > lst[i] and i>0:
            lst[i-1], lst[i] = lst[i], lst[i-1]
            i -= 1
    return lst

insertion_sort([7,1,3,5,8,4])
````

# Modified Insertion Sort:

````python
def insertion_sort(lst):
    for k in range(1,len(lst)):
        i = k         #item = lst[i]
        
        while i > 0:
            if lst[i-1] > lst[i]:
                lst[i-1], lst[i] = lst[i], lst[i-1]
                i -= 1
            else:
                break
    return lst

insertion_sort([7,1,3,5,8,4])
````
