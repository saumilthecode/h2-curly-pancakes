### Q1 Basic Skill 1a : Split a List into Two

```python
def split(seq):              # return a tuple
    if len(seq) >1:
        l = len(seq)//2
        return(seq[:l],seq[l:])
    else:
        return seq



print(split([5, 2, 1, 8, 9])==([5, 2], [1, 8, 9]))
split([5, 2, 1, 8, 9])
```

### Q2 Basic Skill 1b : Recursively Split a List into Two

```python
def split(seq):              # return a tuple
    if len(seq) > 1:
        middle = len(seq) // 2
        return (split(seq[:middle]), split(seq[middle:]))
    else:
        return seq


print(split([5, 2, 1, 8, 9])==(([5], [2]), ([1], ([8], [9]))))
```

### Q4 Basic Skill 2a : Merge 2 Lists each with 1 element

```python
def merge(left, right):    # if left and right only has 1 element each
    if left[0] < right[0]:
        return [left[0],right[0]]
    else:
        return [right[0],left[0]]



print(merge([8], [9])==[8, 9])
print(merge([5], [2])==[2, 5])
merge([8], [9])
```

### Q5 Basic Skill 2b : Merge 2 Lists (1 list with 1 element and the other is a SORTED list)

```python
def merge(left, right):  # if left or right has more than 1 element, but not both
    merged = []

    while left and right:
        if left[0] < right[0]:
            merged.append(left.pop(0))
        else:
            merged.append(right.pop(0))

    merged += left
    merged += right

    return merged

print(merge([1], [8,9])==[1, 8, 9])
print(merge([8,9], [1])==[1, 8, 9])
```

### Q6 Basic Skill 2c : Merge 2 Sorted Lists of any no. of elements 

```python
def merge(left, right):
    merged = []

    while left and right:
        if left[0] < right[0]:
            merged.append(left.pop(0))
        else:
            merged.append(right.pop(0))

    merged += left
    merged += right

    return merged



print(merge([2, 5],[1,8,9])==[1, 2, 5, 8, 9])
print(merge([2,6,7], [9,10])==[2, 6, 7, 9, 10])
```
